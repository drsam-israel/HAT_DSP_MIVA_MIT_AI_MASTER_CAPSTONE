```python

# ============================================================
# HAT-DSP
# D00.2 — RUNTIME ENVIRONMENT AND DEPENDENCY MANAGEMENT
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
from importlib import import_module, metadata

import sys
import platform
import subprocess
import json
import csv
import hashlib
import os

from google.colab import drive

# ------------------------------------------------------------
# 1. Establish persistent workspace connectivity
# ------------------------------------------------------------

if not Path("/content/drive/MyDrive").is_dir():
    drive.mount("/content/drive")

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/"
    "HAT_DSP_MIVA_MIT_AI_CAPSTONE"
)

GOVERNANCE_DIR = PROJECT_ROOT / "00_project_governance"
LOG_DIR = PROJECT_ROOT / "23_logs"

CONFIG_PATH = (
    GOVERNANCE_DIR / "project_config_v1.json"
)

assert PROJECT_ROOT.is_dir(), "Project workspace not found."
assert CONFIG_PATH.is_file(), "Project configuration not found."
assert LOG_DIR.is_dir(), "Project log directory not found."

EXPECTED_CONFIG_SHA256 = (
    "4d4a5ca9653713c7160deb95300608cf87"
    "fa92d5c92c6903590cbbc9ba980d5c"
)

ACTUAL_CONFIG_SHA256 = hashlib.sha256(
    CONFIG_PATH.read_bytes()
).hexdigest()

assert ACTUAL_CONFIG_SHA256 == EXPECTED_CONFIG_SHA256, (
    "Configuration fingerprint mismatch. "
    "Execution stopped pending integrity investigation."
)

with CONFIG_PATH.open("r", encoding="utf-8") as file:
    PROJECT_CONFIG = json.load(file)

# ------------------------------------------------------------
# 2. Define environment evidence directory
# ------------------------------------------------------------

ENVIRONMENT_DIR = GOVERNANCE_DIR / "environment"

ENVIRONMENT_DIR.mkdir(exist_ok=True)

RUN_TIMESTAMP = datetime.now(timezone.utc).strftime(
    "%Y%m%dT%H%M%S_%fZ"
)

# ------------------------------------------------------------
# 3. Define core project dependencies
# ------------------------------------------------------------

CORE_DEPENDENCIES = {
    "numpy": "numpy",
    "pandas": "pandas",
    "scipy": "scipy",
    "scikit-learn": "sklearn",
    "xgboost": "xgboost",
    "shap": "shap",
    "matplotlib": "matplotlib",
    "seaborn": "seaborn",
    "plotly": "plotly",
    "streamlit": "streamlit",
    "joblib": "joblib",
    "requests": "requests",
    "statsmodels": "statsmodels",
    "pytest": "pytest",
    "psutil": "psutil"
}

print("=" * 75)
print("HAT-DSP | D00.2")
print("RUNTIME ENVIRONMENT AND DEPENDENCY MANAGEMENT")
print("=" * 75)

print("\nConfiguration integrity: PASS")

# ------------------------------------------------------------
# 4. Identify missing dependencies
# ------------------------------------------------------------

MISSING_DEPENDENCIES = []

for package in CORE_DEPENDENCIES:

    try:
        metadata.version(package)

    except metadata.PackageNotFoundError:
        MISSING_DEPENDENCIES.append(package)

print("\nDEPENDENCY AVAILABILITY")
print("-" * 75)

if MISSING_DEPENDENCIES:

    print("Missing packages:", MISSING_DEPENDENCIES)
    print("Installing missing dependencies only...")

    subprocess.run(
        [
            sys.executable,
            "-m",
            "pip",
            "install",
            *MISSING_DEPENDENCIES
        ],
        check=True
    )

else:

    print("All core dependencies are already installed.")

# ------------------------------------------------------------
# 5. Verify dependency versions and importability
# ------------------------------------------------------------

DEPENDENCY_RECORDS = []

print("\nCORE DEPENDENCY VERIFICATION")
print("-" * 75)

for package, module_name in CORE_DEPENDENCIES.items():

    installed_version = None
    import_error = None

    try:

        installed_version = metadata.version(package)
        import_module(module_name)

        status = "PASS"

    except Exception as error:

        status = "FAIL"
        import_error = str(error)

    DEPENDENCY_RECORDS.append({
        "package": package,
        "python_module": module_name,
        "installed_version": installed_version,
        "verification_status": status,
        "import_error": import_error
    })

    print(
        f"{package:<20}"
        f"{str(installed_version):<20}"
        f"{status}"
    )

# ------------------------------------------------------------
# 6. Capture runtime environment information
# ------------------------------------------------------------

import psutil

RUNTIME_MANIFEST = {
    "project": "HAT-DSP",
    "stage": "D00.2",
    "captured_utc": datetime.now(timezone.utc).isoformat(),
    "development_environment": "Google Colab",
    "python_version": platform.python_version(),
    "python_implementation": platform.python_implementation(),
    "python_executable": sys.executable,
    "operating_system": platform.system(),
    "operating_system_release": platform.release(),
    "machine_architecture": platform.machine(),
    "logical_cpu_count": os.cpu_count(),
    "available_runtime_ram_gib": round(
        psutil.virtual_memory().total / (1024 ** 3), 2
    ),
    "configuration_version": PROJECT_CONFIG[
        "configuration_version"
    ],
    "configuration_sha256": ACTUAL_CONFIG_SHA256,
    "random_seed": PROJECT_CONFIG["random_seed"],
    "core_dependency_count": len(CORE_DEPENDENCIES),
    "newly_installed_dependencies": MISSING_DEPENDENCIES
}

# ------------------------------------------------------------
# 7. Define versioned artifact paths
# ------------------------------------------------------------

MANIFEST_PATH = (
    ENVIRONMENT_DIR
    / f"runtime_manifest_{RUN_TIMESTAMP}.json"
)

DEPENDENCY_PATH = (
    ENVIRONMENT_DIR
    / f"dependency_register_{RUN_TIMESTAMP}.csv"
)

REQUIREMENTS_PATH = (
    ENVIRONMENT_DIR
    / f"requirements_core_{RUN_TIMESTAMP}.txt"
)

FREEZE_PATH = (
    ENVIRONMENT_DIR
    / f"environment_freeze_{RUN_TIMESTAMP}.txt"
)

# ------------------------------------------------------------
# 8. Persist runtime environment manifest
# ------------------------------------------------------------

with MANIFEST_PATH.open("x", encoding="utf-8") as file:
    json.dump(RUNTIME_MANIFEST, file, indent=4)

# ------------------------------------------------------------
# 9. Persist dependency verification register
# ------------------------------------------------------------

with DEPENDENCY_PATH.open(
    "x", newline="", encoding="utf-8"
) as file:

    writer = csv.DictWriter(
        file,
        fieldnames=[
            "package",
            "python_module",
            "installed_version",
            "verification_status",
            "import_error"
        ]
    )

    writer.writeheader()
    writer.writerows(DEPENDENCY_RECORDS)

# ------------------------------------------------------------
# 10. Generate version-pinned core requirements
# ------------------------------------------------------------

REQUIREMENT_LINES = [
    f"{record['package']}=={record['installed_version']}"
    for record in DEPENDENCY_RECORDS
    if record["verification_status"] == "PASS"
]

with REQUIREMENTS_PATH.open("x", encoding="utf-8") as file:

    file.write(
        "\n".join(sorted(REQUIREMENT_LINES)) + "\n"
    )

# ------------------------------------------------------------
# 11. Capture complete installed-package snapshot
# ------------------------------------------------------------

FREEZE_RESULT = subprocess.run(
    [sys.executable, "-m", "pip", "freeze", "--all"],
    capture_output=True,
    text=True,
    check=True
)

with FREEZE_PATH.open("x", encoding="utf-8") as file:
    file.write(FREEZE_RESULT.stdout)

# ------------------------------------------------------------
# 12. Verify dependency consistency
# ------------------------------------------------------------

PIP_CHECK = subprocess.run(
    [sys.executable, "-m", "pip", "check"],
    capture_output=True,
    text=True
)

PIP_CHECK_STATUS = (
    "PASS" if PIP_CHECK.returncode == 0
    else "REVIEW REQUIRED"
)

PIP_CHECK_OUTPUT = (
    PIP_CHECK.stdout.strip()
    + "\n"
    + PIP_CHECK.stderr.strip()
).strip()

# ------------------------------------------------------------
# 13. Verify environment artifacts
# ------------------------------------------------------------

ENVIRONMENT_ARTIFACTS = [
    MANIFEST_PATH,
    DEPENDENCY_PATH,
    REQUIREMENTS_PATH,
    FREEZE_PATH
]

ARTIFACT_HASHES = {
    path.name: hashlib.sha256(
        path.read_bytes()
    ).hexdigest()
    for path in ENVIRONMENT_ARTIFACTS
}

with MANIFEST_PATH.open("r", encoding="utf-8") as file:
    RELOADED_MANIFEST = json.load(file)

with DEPENDENCY_PATH.open("r", encoding="utf-8") as file:
    RELOADED_DEPENDENCIES = list(csv.DictReader(file))

RELOADED_REQUIREMENTS = (
    REQUIREMENTS_PATH.read_text(
        encoding="utf-8"
    ).splitlines()
)

VERIFICATION_RESULTS = {
    "project_configuration_integrity": (
        ACTUAL_CONFIG_SHA256 == EXPECTED_CONFIG_SHA256
    ),

    "core_dependencies_importable": all(
        record["verification_status"] == "PASS"
        for record in DEPENDENCY_RECORDS
    ),

    "dependency_versions_recorded": all(
        bool(record["installed_version"])
        for record in DEPENDENCY_RECORDS
    ),

    "runtime_manifest_reload_integrity": (
        RELOADED_MANIFEST == RUNTIME_MANIFEST
    ),

    "dependency_register_complete": (
        len(RELOADED_DEPENDENCIES) ==
        len(CORE_DEPENDENCIES)
    ),

    "core_requirements_complete": (
        len(RELOADED_REQUIREMENTS) ==
        len(CORE_DEPENDENCIES)
    ),

    "environment_snapshot_nonempty": (
        FREEZE_PATH.stat().st_size > 0
    ),

    "environment_artifacts_persisted": all(
        path.is_file() and path.stat().st_size > 0
        for path in ENVIRONMENT_ARTIFACTS
    )
}

TOTAL_CHECKS = len(VERIFICATION_RESULTS)
PASSED_CHECKS = sum(VERIFICATION_RESULTS.values())
FAILED_CHECKS = TOTAL_CHECKS - PASSED_CHECKS

if FAILED_CHECKS > 0:

    ENVIRONMENT_STATUS = "FAIL"

elif PIP_CHECK.returncode != 0:

    ENVIRONMENT_STATUS = "REVIEW REQUIRED"

else:

    ENVIRONMENT_STATUS = "PASS"

# ------------------------------------------------------------
# 14. Persist environment verification report
# ------------------------------------------------------------

VERIFICATION_REPORT = {
    "project": "HAT-DSP",
    "stage": "D00.2",
    "verified_utc": datetime.now(timezone.utc).isoformat(),
    "verification_results": VERIFICATION_RESULTS,
    "total_checks": TOTAL_CHECKS,
    "passed_checks": PASSED_CHECKS,
    "failed_checks": FAILED_CHECKS,
    "pip_check_status": PIP_CHECK_STATUS,
    "pip_check_output": PIP_CHECK_OUTPUT,
    "artifact_sha256": ARTIFACT_HASHES,
    "environment_status": ENVIRONMENT_STATUS,
    "fresh_runtime_reproduction_verified": False
}

REPORT_PATH = (
    LOG_DIR
    / f"D00_2_environment_{RUN_TIMESTAMP}.json"
)

with REPORT_PATH.open("x", encoding="utf-8") as file:
    json.dump(VERIFICATION_REPORT, file, indent=4)

# ------------------------------------------------------------
# 15. Display environment assessment results
# ------------------------------------------------------------

print("\nRUNTIME ENVIRONMENT")
print("-" * 75)

print("Python version       :", RUNTIME_MANIFEST["python_version"])
print("Operating system     :", RUNTIME_MANIFEST["operating_system"])
print("Architecture         :", RUNTIME_MANIFEST["machine_architecture"])
print("Logical CPUs         :", RUNTIME_MANIFEST["logical_cpu_count"])
print("Runtime RAM (GiB)    :", RUNTIME_MANIFEST["available_runtime_ram_gib"])
print("Random seed          :", RUNTIME_MANIFEST["random_seed"])

print("\nENVIRONMENT VERIFICATION")
print("-" * 75)

for check, result in VERIFICATION_RESULTS.items():

    print(
        f"{check.replace('_', ' ').title():<48}: "
        f"{'PASS' if result else 'FAIL'}"
    )

print("\nDEPENDENCY CONSISTENCY")
print("-" * 75)

print("Pip check status:", PIP_CHECK_STATUS)
print(PIP_CHECK_OUTPUT)

print("\nENVIRONMENT SUMMARY")
print("-" * 75)

print("Core dependencies     :", len(CORE_DEPENDENCIES))

print(
    "Verified imports      :",
    sum(
        record["verification_status"] == "PASS"
        for record in DEPENDENCY_RECORDS
    )
)

print("New packages installed:", len(MISSING_DEPENDENCIES))
print("Total checks          :", TOTAL_CHECKS)
print("Passed checks         :", PASSED_CHECKS)
print("Failed checks         :", FAILED_CHECKS)

print("\nPERSISTED ENVIRONMENT ARTIFACTS")
print("-" * 75)

for path in ENVIRONMENT_ARTIFACTS:
    print(path.name)

print("\nEnvironment directory:")
print(ENVIRONMENT_DIR)

print("\nVerification report:")
print(REPORT_PATH)

print("=" * 75)
print("D00.2 ENVIRONMENT STATUS:", ENVIRONMENT_STATUS)
print("=" * 75)

assert FAILED_CHECKS == 0, (
    "Required environment verification checks failed."
)
```

### D00.2A — Targeted Dependency Remediation

#### Identified Issue

The initial environment verification identified one missing dependency required by IPython:

- IPython requires Jedi.
- All 15 core project dependencies imported successfully.
- All eight foundational environment verification checks passed.
- Overall environment acceptance remains pending dependency consistency verification.

#### Remediation Strategy

Install the missing Jedi dependency using the version requirement declared by IPython, without unnecessarily upgrading existing core project dependencies.

Following remediation:

1. Verify Jedi installation and importability.
2. Re-execute the dependency consistency check.
3. Confirm that existing core dependency versions remain unchanged.
4. Preserve a timestamped remediation record.
5. Refresh the environment baseline after successful remediation.

#### Acceptance Criteria

The remediation shall be accepted when Jedi is available and importable, the dependency consistency check passes, existing core dependency versions remain unchanged, and remediation evidence is successfully persisted.


```python

# ============================================================
# HAT-DSP
# D00.2A — TARGETED DEPENDENCY REMEDIATION
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
from importlib import import_module, metadata
from packaging.requirements import Requirement
from packaging.utils import canonicalize_name

import sys
import subprocess
import json
import csv
import hashlib

# ------------------------------------------------------------
# 1. Establish project artifact locations
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/"
    "HAT_DSP_MIVA_MIT_AI_CAPSTONE"
)

ENVIRONMENT_DIR = (
    PROJECT_ROOT
    / "00_project_governance"
    / "environment"
)

LOG_DIR = PROJECT_ROOT / "23_logs"

assert ENVIRONMENT_DIR.is_dir()
assert LOG_DIR.is_dir()

BASELINE_REGISTER = (
    ENVIRONMENT_DIR
    / "dependency_register_20261118T191323_062368Z.csv"
)

assert BASELINE_REGISTER.is_file(), (
    "D00.2 dependency baseline not found."
)

# ------------------------------------------------------------
# 2. Load verified core dependency baseline
# ------------------------------------------------------------

with BASELINE_REGISTER.open(
    "r", encoding="utf-8"
) as file:
    BASELINE_RECORDS = list(csv.DictReader(file))

BASELINE_VERSIONS = {
    record["package"]: record["installed_version"]
    for record in BASELINE_RECORDS
}

assert len(BASELINE_VERSIONS) == 15

print("=" * 75)
print("HAT-DSP | D00.2A")
print("TARGETED DEPENDENCY REMEDIATION")
print("=" * 75)

# ------------------------------------------------------------
# 3. Identify IPython's declared Jedi requirement
# ------------------------------------------------------------

IPYTHON_REQUIREMENTS = (
    metadata.requires("ipython") or []
)

JEDI_REQUIREMENTS = []

for requirement_text in IPYTHON_REQUIREMENTS:

    requirement = Requirement(requirement_text)

    if canonicalize_name(requirement.name) == "jedi":

        if (
            requirement.marker is None
            or requirement.marker.evaluate()
        ):
            JEDI_REQUIREMENTS.append(requirement)

assert JEDI_REQUIREMENTS, (
    "No applicable Jedi requirement found in "
    "IPython package metadata."
)

JEDI_INSTALL_SPEC = (
    "jedi" + ",".join(
        str(requirement.specifier)
        for requirement in JEDI_REQUIREMENTS
        if str(requirement.specifier)
    )
)

print("\nIDENTIFIED REQUIREMENT")
print("-" * 75)

print("IPython version :", metadata.version("ipython"))
print("Jedi requirement:", JEDI_INSTALL_SPEC)

# ------------------------------------------------------------
# 4. Install Jedi only if required
# ------------------------------------------------------------

try:

    EXISTING_JEDI_VERSION = metadata.version("jedi")

except metadata.PackageNotFoundError:

    EXISTING_JEDI_VERSION = None

JEDI_COMPATIBLE = (
    EXISTING_JEDI_VERSION is not None
    and all(
        requirement.specifier.contains(
            EXISTING_JEDI_VERSION,
            prereleases=True
        )
        for requirement in JEDI_REQUIREMENTS
    )
)

if JEDI_COMPATIBLE:

    print("\nJedi is already installed and compatible.")

else:

    print("\nInstalling declared Jedi dependency...")

    subprocess.run(
        [
            sys.executable,
            "-m",
            "pip",
            "install",
            JEDI_INSTALL_SPEC
        ],
        check=True
    )

# ------------------------------------------------------------
# 5. Verify Jedi installation and import
# ------------------------------------------------------------

JEDI_VERSION = metadata.version("jedi")

try:

    import_module("jedi")
    JEDI_IMPORT_CHECK = True

except Exception:

    JEDI_IMPORT_CHECK = False

JEDI_REQUIREMENT_CHECK = all(
    requirement.specifier.contains(
        JEDI_VERSION,
        prereleases=True
    )
    for requirement in JEDI_REQUIREMENTS
)

# ------------------------------------------------------------
# 6. Check preservation of core dependency versions
# ------------------------------------------------------------

CURRENT_VERSIONS = {
    package: metadata.version(package)
    for package in BASELINE_VERSIONS
}

VERSION_CHANGES = {
    package: {
        "baseline": BASELINE_VERSIONS[package],
        "current": CURRENT_VERSIONS[package]
    }
    for package in BASELINE_VERSIONS
    if (
        BASELINE_VERSIONS[package]
        != CURRENT_VERSIONS[package]
    )
}

CORE_VERSION_CHECK = len(VERSION_CHANGES) == 0

# Verify core imports in a fresh Python subprocess
# rather than relying on already loaded modules.

MODULE_NAMES = [
    record["python_module"]
    for record in BASELINE_RECORDS
]

IMPORT_TEST_CODE = (
    "import importlib\n"
    f"modules = {MODULE_NAMES!r}\n"
    "for module in modules:\n"
    "    importlib.import_module(module)\n"
    "print('ALL CORE IMPORTS PASSED')\n"
)

CORE_IMPORT_TEST = subprocess.run(
    [sys.executable, "-c", IMPORT_TEST_CODE],
    capture_output=True,
    text=True
)

CORE_IMPORT_CHECK = CORE_IMPORT_TEST.returncode == 0

# ------------------------------------------------------------
# 7. Re-execute dependency consistency verification
# ------------------------------------------------------------

PIP_CHECK = subprocess.run(
    [sys.executable, "-m", "pip", "check"],
    capture_output=True,
    text=True
)

PIP_CHECK_OUTPUT = (
    PIP_CHECK.stdout.strip()
    + "\n"
    + PIP_CHECK.stderr.strip()
).strip()

PIP_CHECK_PASSED = PIP_CHECK.returncode == 0

# ------------------------------------------------------------
# 8. Evaluate remediation acceptance criteria
# ------------------------------------------------------------

VERIFICATION_RESULTS = {
    "jedi_importable": JEDI_IMPORT_CHECK,
    "jedi_requirement_satisfied": JEDI_REQUIREMENT_CHECK,
    "core_dependency_versions_preserved": CORE_VERSION_CHECK,
    "fresh_process_core_imports": CORE_IMPORT_CHECK,
    "dependency_consistency": PIP_CHECK_PASSED
}

TOTAL_CHECKS = len(VERIFICATION_RESULTS)
PASSED_CHECKS = sum(VERIFICATION_RESULTS.values())
FAILED_CHECKS = TOTAL_CHECKS - PASSED_CHECKS

REMEDIATION_STATUS = (
    "PASS" if FAILED_CHECKS == 0
    else "REVIEW REQUIRED"
)

# ------------------------------------------------------------
# 9. Persist remediation evidence
# ------------------------------------------------------------

RUN_TIMESTAMP = datetime.now(timezone.utc).strftime(
    "%Y%m%dT%H%M%S_%fZ"
)

REMEDIATION_REPORT = {
    "project": "HAT-DSP",
    "stage": "D00.2A",
    "verified_utc": datetime.now(timezone.utc).isoformat(),
    "identified_issue": "Missing IPython Jedi dependency",
    "ipython_version": metadata.version("ipython"),
    "jedi_requirement": JEDI_INSTALL_SPEC,
    "jedi_version": JEDI_VERSION,
    "baseline_register": BASELINE_REGISTER.name,
    "baseline_register_sha256": hashlib.sha256(
        BASELINE_REGISTER.read_bytes()
    ).hexdigest(),
    "core_dependencies": len(BASELINE_VERSIONS),
    "core_dependency_version_changes": VERSION_CHANGES,
    "verification_results": VERIFICATION_RESULTS,
    "core_import_test_output": CORE_IMPORT_TEST.stdout,
    "core_import_test_errors": CORE_IMPORT_TEST.stderr,
    "pip_check_output": PIP_CHECK_OUTPUT,
    "total_checks": TOTAL_CHECKS,
    "passed_checks": PASSED_CHECKS,
    "failed_checks": FAILED_CHECKS,
    "remediation_status": REMEDIATION_STATUS
}

REPORT_PATH = (
    LOG_DIR
    / f"D00_2A_remediation_{RUN_TIMESTAMP}.json"
)

with REPORT_PATH.open("x", encoding="utf-8") as file:
    json.dump(REMEDIATION_REPORT, file, indent=4)

# Verify persisted remediation report.

with REPORT_PATH.open("r", encoding="utf-8") as file:
    RELOADED_REPORT = json.load(file)

assert RELOADED_REPORT == REMEDIATION_REPORT

# ------------------------------------------------------------
# 10. Display remediation assessment
# ------------------------------------------------------------

print("\nREMEDIATION VERIFICATION")
print("-" * 75)

print("Jedi installed version:", JEDI_VERSION)

for check, result in VERIFICATION_RESULTS.items():

    print(
        f"{check.replace('_', ' ').title():<48}: "
        f"{'PASS' if result else 'FAIL'}"
    )

print("\nDEPENDENCY CONSISTENCY")
print("-" * 75)
print(PIP_CHECK_OUTPUT)

print("\nREMEDIATION SUMMARY")
print("-" * 75)

print("Core dependencies preserved:", len(BASELINE_VERSIONS))
print("Core version changes        :", len(VERSION_CHANGES))
print("Total verification checks   :", TOTAL_CHECKS)
print("Passed checks               :", PASSED_CHECKS)
print("Failed checks               :", FAILED_CHECKS)
print("Remediation report integrity: PASS")

if VERSION_CHANGES:
    print("\nUnexpected version changes:")
    print(json.dumps(VERSION_CHANGES, indent=4))

if not CORE_IMPORT_CHECK:
    print("\nFresh-process import errors:")
    print(CORE_IMPORT_TEST.stderr)

print("\nREMEDIATION ARTIFACT")
print(REPORT_PATH)

print("=" * 75)
print("D00.2A REMEDIATION STATUS:", REMEDIATION_STATUS)
print("=" * 75)
```

Successful environment baseline acceptance establishes a documented development environment but does not constitute proof of complete fresh-runtime reproducibility.


```python

# ============================================================
# HAT-DSP
# D00.2B — ENVIRONMENT BASELINE CONSOLIDATION AND FREEZE
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
from importlib import metadata

import sys
import platform
import subprocess
import hashlib
import json
import csv

# ------------------------------------------------------------
# 1. Establish project artifact locations
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/"
    "HAT_DSP_MIVA_MIT_AI_CAPSTONE"
)

GOVERNANCE_DIR = PROJECT_ROOT / "00_project_governance"
ENVIRONMENT_DIR = GOVERNANCE_DIR / "environment"
LOG_DIR = PROJECT_ROOT / "23_logs"

CONFIG_PATH = (
    GOVERNANCE_DIR / "project_config_v1.json"
)

INITIAL_REPORT_PATH = (
    LOG_DIR
    / "D00_2_environment_20261118T191323_062368Z.json"
)

REMEDIATION_REPORT_PATH = (
    LOG_DIR
    / "D00_2A_remediation_20261118T191747_332499Z.json"
)

BASELINE_REGISTER_PATH = (
    ENVIRONMENT_DIR
    / "dependency_register_20261118T191323_062368Z.csv"
)

for path in [
    CONFIG_PATH,
    INITIAL_REPORT_PATH,
    REMEDIATION_REPORT_PATH,
    BASELINE_REGISTER_PATH
]:
    assert path.is_file(), f"Required artifact missing: {path}"

def sha256_file(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()

def read_json(path):
    return json.loads(path.read_text(encoding="utf-8"))

def write_json_exclusive(path, content):
    with path.open("x", encoding="utf-8") as file:
        json.dump(content, file, indent=4)

print("=" * 75)
print("HAT-DSP | D00.2B")
print("ENVIRONMENT BASELINE CONSOLIDATION AND FREEZE")
print("=" * 75)

# ------------------------------------------------------------
# 2. Verify prerequisite evidence
# ------------------------------------------------------------

INITIAL_REPORT = read_json(INITIAL_REPORT_PATH)
REMEDIATION_REPORT = read_json(REMEDIATION_REPORT_PATH)

assert INITIAL_REPORT["failed_checks"] == 0

assert REMEDIATION_REPORT["remediation_status"] == "PASS"
assert REMEDIATION_REPORT["failed_checks"] == 0
assert REMEDIATION_REPORT["core_dependency_version_changes"] == {}

EXPECTED_CONFIG_HASH = (
    "4d4a5ca9653713c7160deb95300608cf87"
    "fa92d5c92c6903590cbbc9ba980d5c"
)

CONFIG_INTEGRITY = (
    sha256_file(CONFIG_PATH) == EXPECTED_CONFIG_HASH
)

INITIAL_ARTIFACT_RESULTS = {
    filename: (
        (ENVIRONMENT_DIR / filename).is_file()
        and sha256_file(ENVIRONMENT_DIR / filename)
        == expected_hash
    )
    for filename, expected_hash
    in INITIAL_REPORT["artifact_sha256"].items()
}

INITIAL_ARTIFACT_INTEGRITY = all(
    INITIAL_ARTIFACT_RESULTS.values()
)

REMEDIATION_LINK_INTEGRITY = (
    BASELINE_REGISTER_PATH.name
    == REMEDIATION_REPORT["baseline_register"]
    and sha256_file(BASELINE_REGISTER_PATH)
    == REMEDIATION_REPORT["baseline_register_sha256"]
)

assert CONFIG_INTEGRITY, "Configuration integrity failure."

assert INITIAL_ARTIFACT_INTEGRITY, (
    "Initial environment artifact integrity failure."
)

assert REMEDIATION_LINK_INTEGRITY, (
    "Remediation evidence linkage failure."
)

print("\nPREREQUISITE EVIDENCE")
print("-" * 75)

print("Project configuration integrity : PASS")
print("Initial environment artifacts   : PASS")
print("Remediation evidence linkage    : PASS")

# ------------------------------------------------------------
# 3. Load approved dependency specifications
# ------------------------------------------------------------

with BASELINE_REGISTER_PATH.open(
    "r", encoding="utf-8"
) as file:
    CORE_RECORDS = list(csv.DictReader(file))

assert len(CORE_RECORDS) == 15

assert all(
    record["verification_status"] == "PASS"
    for record in CORE_RECORDS
)

CONSOLIDATED_RECORDS = []

for record in CORE_RECORDS:

    CONSOLIDATED_RECORDS.append({
        "package": record["package"],
        "python_module": record["python_module"],
        "approved_version": record["installed_version"],
        "dependency_category": "CORE_PROJECT"
    })

CONSOLIDATED_RECORDS.append({
    "package": "jedi",
    "python_module": "jedi",
    "approved_version": REMEDIATION_REPORT["jedi_version"],
    "dependency_category": "RUNTIME_SUPPORT"
})

# ------------------------------------------------------------
# 4. Verify installed dependency versions
# ------------------------------------------------------------

print("\nCONSOLIDATED DEPENDENCY VERIFICATION")
print("-" * 75)

for record in CONSOLIDATED_RECORDS:

    try:
        installed_version = metadata.version(record["package"])

    except metadata.PackageNotFoundError:
        installed_version = None

    record["installed_version"] = installed_version

    record["version_status"] = (
        "PASS"
        if installed_version == record["approved_version"]
        else "FAIL"
    )

    print(
        f"{record['package']:<20}"
        f"{str(installed_version):<18}"
        f"{record['version_status']}"
    )

VERSION_INTEGRITY = all(
    record["version_status"] == "PASS"
    for record in CONSOLIDATED_RECORDS
)

# ------------------------------------------------------------
# 5. Verify imports in a fresh Python process
# ------------------------------------------------------------

MODULE_NAMES = [
    record["python_module"]
    for record in CONSOLIDATED_RECORDS
]

IMPORT_CODE = (
    "import importlib\n"
    f"modules = {MODULE_NAMES!r}\n"
    "for module in modules:\n"
    "    importlib.import_module(module)\n"
    "print('All approved dependencies imported successfully.')\n"
)

IMPORT_TEST = subprocess.run(
    [sys.executable, "-c", IMPORT_CODE],
    capture_output=True,
    text=True
)

IMPORT_STATUS = IMPORT_TEST.returncode == 0

# ------------------------------------------------------------
# 6. Reconfirm dependency consistency
# ------------------------------------------------------------

PIP_CHECK = subprocess.run(
    [sys.executable, "-m", "pip", "check"],
    capture_output=True,
    text=True
)

CONSISTENCY_STATUS = PIP_CHECK.returncode == 0

PIP_CHECK_OUTPUT = (
    PIP_CHECK.stdout.strip()
    + "\n"
    + PIP_CHECK.stderr.strip()
).strip()

PRE_FREEZE_CHECKS = {
    "configuration_integrity": CONFIG_INTEGRITY,
    "initial_artifact_integrity": INITIAL_ARTIFACT_INTEGRITY,
    "remediation_evidence_linkage": REMEDIATION_LINK_INTEGRITY,
    "approved_dependency_versions_preserved": VERSION_INTEGRITY,
    "fresh_process_imports": IMPORT_STATUS,
    "dependency_consistency": CONSISTENCY_STATUS
}

print("\nPRE-FREEZE VERIFICATION")
print("-" * 75)

for check, result in PRE_FREEZE_CHECKS.items():
    print(
        f"{check.replace('_', ' ').title():<48}: "
        f"{'PASS' if result else 'FAIL'}"
    )

print("\nDependency consistency output:")
print(PIP_CHECK_OUTPUT)

if not IMPORT_STATUS:
    print(IMPORT_TEST.stderr)

assert all(PRE_FREEZE_CHECKS.values()), (
    "Environment freeze blocked: "
    "one or more pre-freeze checks failed."
)

# ------------------------------------------------------------
# 7. Establish baseline directory and version
# ------------------------------------------------------------

BASELINE_VERSION = "1.0.0"

FREEZE_DIR = (
    ENVIRONMENT_DIR / f"baseline_v{BASELINE_VERSION}"
)

assert not FREEZE_DIR.exists(), (
    "Environment baseline already exists. "
    "Overwriting is not permitted."
)

FREEZE_DIR.mkdir()

REQUIREMENTS_PATH = FREEZE_DIR / "requirements.txt"

REGISTER_PATH = (
    FREEZE_DIR / "dependency_register.csv"
)

RUNTIME_PATH = (
    FREEZE_DIR / "runtime_manifest.json"
)

SNAPSHOT_PATH = (
    FREEZE_DIR / "environment_freeze.txt"
)

INTEGRITY_PATH = (
    FREEZE_DIR / "integrity_manifest.json"
)

# ------------------------------------------------------------
# 8. Persist approved project requirements
# ------------------------------------------------------------

REQUIREMENT_LINES = sorted([
    f"{record['package']}=={record['approved_version']}"
    for record in CONSOLIDATED_RECORDS
])

with REQUIREMENTS_PATH.open("x", encoding="utf-8") as file:

    file.write(
        "\n".join(REQUIREMENT_LINES) + "\n"
    )

# ------------------------------------------------------------
# 9. Persist consolidated dependency register
# ------------------------------------------------------------

with REGISTER_PATH.open(
    "x", newline="", encoding="utf-8"
) as file:

    writer = csv.DictWriter(
        file,
        fieldnames=[
            "package",
            "python_module",
            "approved_version",
            "dependency_category",
            "installed_version",
            "version_status"
        ]
    )

    writer.writeheader()
    writer.writerows(CONSOLIDATED_RECORDS)

# ------------------------------------------------------------
# 10. Persist final runtime manifest
# ------------------------------------------------------------

RUNTIME_MANIFEST = {
    "project": "HAT-DSP",
    "stage": "D00.2B",
    "baseline_version": BASELINE_VERSION,
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "development_environment": "Google Colab",
    "python_version": platform.python_version(),
    "python_implementation": platform.python_implementation(),
    "operating_system": platform.system(),
    "operating_system_release": platform.release(),
    "machine_architecture": platform.machine(),
    "configuration_sha256": sha256_file(CONFIG_PATH),
    "core_project_dependencies": 15,
    "runtime_support_dependencies": 1,
    "total_approved_dependencies": len(CONSOLIDATED_RECORDS),
    "dependency_consistency": "PASS",
    "fresh_process_imports": "PASS",
    "initial_environment_report": INITIAL_REPORT_PATH.name,
    "initial_environment_report_sha256": (
        sha256_file(INITIAL_REPORT_PATH)
    ),
    "remediation_report": REMEDIATION_REPORT_PATH.name,
    "remediation_report_sha256": (
        sha256_file(REMEDIATION_REPORT_PATH)
    ),
    "fresh_runtime_reproduction_verified": False
}

write_json_exclusive(
    RUNTIME_PATH,
    RUNTIME_MANIFEST
)

# ------------------------------------------------------------
# 11. Capture final installed-package snapshot
# ------------------------------------------------------------

FREEZE_RESULT = subprocess.run(
    [sys.executable, "-m", "pip", "freeze", "--all"],
    capture_output=True,
    text=True,
    check=True
)

with SNAPSHOT_PATH.open("x", encoding="utf-8") as file:
    file.write(FREEZE_RESULT.stdout)

# ------------------------------------------------------------
# 12. Generate baseline integrity manifest
# ------------------------------------------------------------

BASELINE_ARTIFACTS = [
    REQUIREMENTS_PATH,
    REGISTER_PATH,
    RUNTIME_PATH,
    SNAPSHOT_PATH
]

INTEGRITY_MANIFEST = {
    "project": "HAT-DSP",
    "baseline_version": BASELINE_VERSION,
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "hash_algorithm": "SHA-256",
    "artifacts": {
        path.name: {
            "size_bytes": path.stat().st_size,
            "sha256": sha256_file(path)
        }
        for path in BASELINE_ARTIFACTS
    }
}

write_json_exclusive(
    INTEGRITY_PATH,
    INTEGRITY_MANIFEST
)

# ------------------------------------------------------------
# 13. Verify final baseline artifacts
# ------------------------------------------------------------

RELOADED_INTEGRITY = read_json(INTEGRITY_PATH)

BASELINE_HASH_CHECK = all(
    (FREEZE_DIR / filename).is_file()
    and sha256_file(FREEZE_DIR / filename)
    == artifact["sha256"]
    and (FREEZE_DIR / filename).stat().st_size
    == artifact["size_bytes"]
    for filename, artifact
    in RELOADED_INTEGRITY["artifacts"].items()
)

RELOADED_REQUIREMENTS = (
    REQUIREMENTS_PATH.read_text(
        encoding="utf-8"
    ).splitlines()
)

with REGISTER_PATH.open("r", encoding="utf-8") as file:
    RELOADED_REGISTER = list(csv.DictReader(file))

VERIFICATION_RESULTS = {
    **PRE_FREEZE_CHECKS,

    "requirements_reload_integrity": (
        RELOADED_REQUIREMENTS == REQUIREMENT_LINES
    ),

    "dependency_register_reload_integrity": (
        RELOADED_REGISTER == CONSOLIDATED_RECORDS
    ),

    "runtime_manifest_reload_integrity": (
        read_json(RUNTIME_PATH) == RUNTIME_MANIFEST
    ),

    "environment_snapshot_nonempty": (
        SNAPSHOT_PATH.stat().st_size > 0
    ),

    "integrity_manifest_reload": (
        RELOADED_INTEGRITY == INTEGRITY_MANIFEST
    ),

    "baseline_artifact_integrity": BASELINE_HASH_CHECK
}

TOTAL_CHECKS = len(VERIFICATION_RESULTS)
PASSED_CHECKS = sum(VERIFICATION_RESULTS.values())
FAILED_CHECKS = TOTAL_CHECKS - PASSED_CHECKS

BASELINE_STATUS = (
    "PASS" if FAILED_CHECKS == 0 else "FAIL"
)

# ------------------------------------------------------------
# 14. Persist final environment acceptance evidence
# ------------------------------------------------------------

RUN_TIMESTAMP = datetime.now(timezone.utc).strftime(
    "%Y%m%dT%H%M%S_%fZ"
)

ACCEPTANCE_REPORT = {
    "project": "HAT-DSP",
    "stage": "D00.2B",
    "parent_stage": "D00.2",
    "verified_utc": datetime.now(timezone.utc).isoformat(),
    "baseline_version": BASELINE_VERSION,
    "baseline_directory": str(FREEZE_DIR),
    "verification_results": VERIFICATION_RESULTS,
    "total_checks": TOTAL_CHECKS,
    "passed_checks": PASSED_CHECKS,
    "failed_checks": FAILED_CHECKS,
    "approved_dependencies": len(CONSOLIDATED_RECORDS),
    "integrity_manifest_sha256": sha256_file(INTEGRITY_PATH),
    "dependency_consistency_output": PIP_CHECK_OUTPUT,
    "environment_baseline_status": BASELINE_STATUS,
    "fresh_runtime_reproduction_verified": False
}

REPORT_PATH = (
    LOG_DIR
    / f"D00_2B_environment_acceptance_{RUN_TIMESTAMP}.json"
)

write_json_exclusive(
    REPORT_PATH,
    ACCEPTANCE_REPORT
)

assert read_json(REPORT_PATH) == ACCEPTANCE_REPORT

# ------------------------------------------------------------
# 15. Display final acceptance results
# ------------------------------------------------------------

print("\nFINAL BASELINE VERIFICATION")
print("-" * 75)

for check, result in VERIFICATION_RESULTS.items():

    print(
        f"{check.replace('_', ' ').title():<48}: "
        f"{'PASS' if result else 'FAIL'}"
    )

print("\nENVIRONMENT BASELINE SUMMARY")
print("-" * 75)

print("Baseline version         :", BASELINE_VERSION)
print("Core project dependencies:", 15)
print("Runtime dependencies     :", 1)
print("Approved dependencies    :", len(CONSOLIDATED_RECORDS))
print("Total verification checks:", TOTAL_CHECKS)
print("Passed checks            :", PASSED_CHECKS)
print("Failed checks            :", FAILED_CHECKS)
print("Baseline artifacts       :", len(BASELINE_ARTIFACTS) + 1)
print("Dependency consistency   : PASS")
print("Acceptance report reload : PASS")
print("Fresh-runtime recovery   : PENDING")

print("\nAPPROVED ENVIRONMENT ARTIFACTS")
print("-" * 75)

for path in BASELINE_ARTIFACTS + [INTEGRITY_PATH]:
    print(f"{path.name:<32} {path.stat().st_size:,} bytes")

print("\nBASELINE DIRECTORY")
print(FREEZE_DIR)

print("\nACCEPTANCE REPORT")
print(REPORT_PATH)

print("=" * 75)
print("D00.2B BASELINE STATUS:", BASELINE_STATUS)
print("=" * 75)

assert FAILED_CHECKS == 0, (
    "Environment baseline acceptance criteria not satisfied."
)
```

```python

# ============================================================
# HAT-DSP
# D00.3B — CONTROLLED CONFIGURATION REVISION
# AND INDEPENDENT GIT INITIALIZATION
# ============================================================

from pathlib import Path
from datetime import datetime, timezone

import hashlib
import subprocess
import json
import requests

# ------------------------------------------------------------
# 1. Establish project locations
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/"
    "HAT_DSP_MIVA_MIT_AI_CAPSTONE"
)

GOVERNANCE_DIR = PROJECT_ROOT / "00_project_governance"
LOG_DIR = PROJECT_ROOT / "23_logs"

ORIGINAL_CONFIG_PATH = (
    GOVERNANCE_DIR / "project_config_v1.json"
)

REVISED_CONFIG_PATH = (
    GOVERNANCE_DIR / "project_config_v1_0_1.json"
)

CHANGE_REGISTER_PATH = (
    GOVERNANCE_DIR / "HAT_DSP_CR_001.json"
)

ENVIRONMENT_BASELINE_DIR = (
    GOVERNANCE_DIR
    / "environment"
    / "baseline_v1.0.0"
)

ENVIRONMENT_ACCEPTANCE_PATH = (
    LOG_DIR
    / "D00_2B_environment_acceptance_20261118T192134_315977Z.json"
)

# Keep the version-controlled workspace separate
# from persistent datasets and research artifacts.

GIT_WORKSPACE = Path(
    "/content/HAT_DSP_MIVA_MIT_AI_MASTER_CAPSTONE"
)

REPOSITORY_OWNER = "drsam-israel"

REPOSITORY_NAME = (
    "HAT_DSP_MIVA_MIT_AI_MASTER_CAPSTONE"
)

REMOTE_URL = (
    f"https://github.com/"
    f"{REPOSITORY_OWNER}/{REPOSITORY_NAME}.git"
)

GITHUB_API_URL = (
    f"https://api.github.com/repos/"
    f"{REPOSITORY_OWNER}/{REPOSITORY_NAME}"
)

def sha256_file(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()

def read_json(path):
    return json.loads(path.read_text(encoding="utf-8"))

def write_json_exclusive(path, content):
    with path.open("x", encoding="utf-8") as file:
        json.dump(content, file, indent=4)

def git_run(*arguments):
    return subprocess.run(
        ["git", "-C", str(GIT_WORKSPACE), *arguments],
        capture_output=True,
        text=True,
        check=True
    ).stdout.strip()

print("=" * 75)
print("HAT-DSP | D00.3B")
print("CONTROLLED CONFIGURATION REVISION AND GIT INITIALIZATION")
print("=" * 75)

# ------------------------------------------------------------
# 2. Verify foundational configuration integrity
# ------------------------------------------------------------

EXPECTED_ORIGINAL_HASH = (
    "4d4a5ca9653713c7160deb95300608cf87"
    "fa92d5c92c6903590cbbc9ba980d5c"
)

assert ORIGINAL_CONFIG_PATH.is_file()

ORIGINAL_CONFIG_HASH = sha256_file(
    ORIGINAL_CONFIG_PATH
)

ORIGINAL_CONFIG_INTEGRITY = (
    ORIGINAL_CONFIG_HASH == EXPECTED_ORIGINAL_HASH
)

assert ORIGINAL_CONFIG_INTEGRITY, (
    "Original configuration integrity failure."
)

ORIGINAL_CONFIG = read_json(
    ORIGINAL_CONFIG_PATH
)

assert ORIGINAL_CONFIG["configuration_version"] == "1.0.0"

# ------------------------------------------------------------
# 3. Verify accepted environment baseline
# ------------------------------------------------------------

ENVIRONMENT_ACCEPTANCE = read_json(
    ENVIRONMENT_ACCEPTANCE_PATH
)

assert (
    ENVIRONMENT_ACCEPTANCE["environment_baseline_status"]
    == "PASS"
)

ENVIRONMENT_INTEGRITY_PATH = (
    ENVIRONMENT_BASELINE_DIR
    / "integrity_manifest.json"
)

ENVIRONMENT_MANIFEST_CHECK = (
    sha256_file(ENVIRONMENT_INTEGRITY_PATH)
    == ENVIRONMENT_ACCEPTANCE["integrity_manifest_sha256"]
)

ENVIRONMENT_INTEGRITY = read_json(
    ENVIRONMENT_INTEGRITY_PATH
)

ENVIRONMENT_ARTIFACT_CHECK = all(
    (ENVIRONMENT_BASELINE_DIR / filename).is_file()
    and sha256_file(
        ENVIRONMENT_BASELINE_DIR / filename
    ) == record["sha256"]
    for filename, record
    in ENVIRONMENT_INTEGRITY["artifacts"].items()
)

ENVIRONMENT_BASELINE_CHECK = (
    ENVIRONMENT_MANIFEST_CHECK
    and ENVIRONMENT_ARTIFACT_CHECK
)

assert ENVIRONMENT_BASELINE_CHECK, (
    "Accepted environment baseline integrity failure."
)

print("\nPREREQUISITE VERIFICATION")
print("-" * 75)
print("Original project configuration : PASS")
print("Accepted environment baseline  : PASS")

# ------------------------------------------------------------
# 4. Verify public GitHub repository identity
# ------------------------------------------------------------

RESPONSE = requests.get(
    GITHUB_API_URL,
    headers={
        "Accept": "application/vnd.github+json"
    },
    timeout=30
)

assert RESPONSE.status_code == 200, (
    f"GitHub repository verification failed: "
    f"HTTP {RESPONSE.status_code}"
)

REPOSITORY_METADATA = RESPONSE.json()

REPOSITORY_CHECKS = {
    "repository_owner": (
        REPOSITORY_METADATA["owner"]["login"]
        == REPOSITORY_OWNER
    ),
    "repository_name": (
        REPOSITORY_METADATA["name"]
        == REPOSITORY_NAME
    ),
    "public_visibility": (
        REPOSITORY_METADATA["private"] is False
        and REPOSITORY_METADATA["visibility"] == "public"
    ),
    "remote_url": (
        REPOSITORY_METADATA["clone_url"]
        == REMOTE_URL
    ),
    "repository_not_archived": (
        REPOSITORY_METADATA["archived"] is False
    ),
    "repository_not_forked": (
        REPOSITORY_METADATA["fork"] is False
    )
}

assert all(REPOSITORY_CHECKS.values()), (
    "Repository identity or visibility mismatch."
)

# Confirm the remote repository has no existing refs.
# This prevents accidentally combining Git histories.

REMOTE_REFS = subprocess.run(
    ["git", "ls-remote", REMOTE_URL],
    capture_output=True,
    text=True,
    check=True,
    timeout=60
).stdout.strip()

REMOTE_EMPTY_CHECK = not bool(REMOTE_REFS)

assert REMOTE_EMPTY_CHECK, (
    "Remote repository contains existing Git references. "
    "Execution stopped pending review."
)

print("\nGITHUB REPOSITORY VERIFICATION")
print("-" * 75)

for check, result in REPOSITORY_CHECKS.items():
    print(
        f"{check.replace('_', ' ').title():<40}: "
        f"{'PASS' if result else 'FAIL'}"
    )

print("Remote repository empty                 : PASS")

# ------------------------------------------------------------
# 5. Prepare controlled configuration revision
# ------------------------------------------------------------

REVISED_CONFIG = {
    **ORIGINAL_CONFIG,
    "configuration_version": "1.0.1",
    "github_repository": REPOSITORY_NAME,
    "github_repository_owner": REPOSITORY_OWNER,
    "github_repository_visibility": "public",
    "github_remote_url": REMOTE_URL,
    "configuration_change_reference": "HAT-DSP-CR-001",
    "parent_configuration_version": "1.0.0",
    "parent_configuration_sha256": ORIGINAL_CONFIG_HASH
}

ALLOWED_CHANGES = {
    "configuration_version",
    "github_repository",
    "github_repository_owner",
    "github_repository_visibility",
    "github_remote_url",
    "configuration_change_reference",
    "parent_configuration_version",
    "parent_configuration_sha256"
}

CHANGED_FIELDS = {
    key: {
        "previous": ORIGINAL_CONFIG.get(key),
        "revised": REVISED_CONFIG.get(key)
    }
    for key in set(ORIGINAL_CONFIG) | set(REVISED_CONFIG)
    if ORIGINAL_CONFIG.get(key) != REVISED_CONFIG.get(key)
}

CONTROLLED_CHANGE_CHECK = (
    set(CHANGED_FIELDS).issubset(ALLOWED_CHANGES)
)

assert CONTROLLED_CHANGE_CHECK, (
    "Unexpected configuration changes detected."
)

assert (
    REVISED_CONFIG["clinical_deployment_authorised"]
    is False
)

# ------------------------------------------------------------
# 6. Persist revised configuration without overwriting
# ------------------------------------------------------------

if REVISED_CONFIG_PATH.exists():

    assert read_json(REVISED_CONFIG_PATH) == REVISED_CONFIG, (
        "Existing revised configuration differs "
        "from the approved revision."
    )

else:

    write_json_exclusive(
        REVISED_CONFIG_PATH,
        REVISED_CONFIG
    )

REVISED_CONFIG_HASH = sha256_file(
    REVISED_CONFIG_PATH
)

REVISED_CONFIG_CHECK = (
    read_json(REVISED_CONFIG_PATH) == REVISED_CONFIG
)

assert REVISED_CONFIG_CHECK

# ------------------------------------------------------------
# 7. Persist controlled change record
# ------------------------------------------------------------

CHANGE_RECORD = {
    "project": "HAT-DSP",
    "change_reference": "HAT-DSP-CR-001",
    "previous_configuration_version": "1.0.0",
    "revised_configuration_version": "1.0.1",
    "previous_configuration_sha256": ORIGINAL_CONFIG_HASH,
    "revised_configuration_sha256": REVISED_CONFIG_HASH,
    "change_category": "Repository identity and visibility",
    "change_authority": "Student-approved decision",
    "changed_fields": CHANGED_FIELDS,
    "academic_scope_impact": "None",
    "clinical_use_case_impact": "None",
    "environment_dependency_impact": "None",
    "environment_baseline_version": "1.0.0",
    "environment_baseline_configuration_reference": (
        ORIGINAL_CONFIG_HASH
    ),
    "remote_repository_id": REPOSITORY_METADATA["id"],
    "clinical_deployment_authorised": False
}

if CHANGE_REGISTER_PATH.exists():

    EXISTING_CHANGE_RECORD = read_json(
        CHANGE_REGISTER_PATH
    )

    assert all(
        EXISTING_CHANGE_RECORD.get(key) == value
        for key, value in CHANGE_RECORD.items()
    ), "Existing configuration change record mismatch."

    CHANGE_RECORD = EXISTING_CHANGE_RECORD

else:

    CHANGE_RECORD["recorded_utc"] = (
        datetime.now(timezone.utc).isoformat()
    )

    write_json_exclusive(
        CHANGE_REGISTER_PATH,
        CHANGE_RECORD
    )

CHANGE_RECORD_CHECK = (
    read_json(CHANGE_REGISTER_PATH) == CHANGE_RECORD
)

# ------------------------------------------------------------
# 8. Verify Git availability
# ------------------------------------------------------------

GIT_VERSION = subprocess.run(
    ["git", "--version"],
    capture_output=True,
    text=True,
    check=True
).stdout.strip()

# ------------------------------------------------------------
# 9. Initialize independent local Git repository
# ------------------------------------------------------------

if GIT_WORKSPACE.exists():

    assert (GIT_WORKSPACE / ".git").is_dir(), (
        "Unexpected existing workspace. "
        "Execution stopped to protect existing files."
    )

    assert not git_run("status", "--porcelain"), (
        "Existing workspace contains uncommitted files."
    )

    assert not git_run("for-each-ref"), (
        "Existing workspace contains Git references."
    )

    EXISTING_REMOTES = git_run("remote").splitlines()

    assert set(EXISTING_REMOTES).issubset({"origin"})

    if "origin" in EXISTING_REMOTES:
        assert git_run(
            "remote", "get-url", "origin"
        ) == REMOTE_URL

    EXECUTION_MODE = "VERIFIED EXISTING EMPTY WORKSPACE"

else:

    GIT_WORKSPACE.mkdir()

    git_run("init", "--initial-branch=main")

    EXECUTION_MODE = "NEW GIT WORKSPACE CREATED"

# ------------------------------------------------------------
# 10. Configure approved remote repository
# ------------------------------------------------------------

if "origin" not in git_run("remote").splitlines():

    git_run(
        "remote",
        "add",
        "origin",
        REMOTE_URL
    )

LOCAL_BRANCH = git_run(
    "symbolic-ref",
    "--short",
    "HEAD"
)

CONFIGURED_REMOTE = git_run(
    "remote",
    "get-url",
    "origin"
)

TRACKED_FILES = git_run("ls-files")

WORKTREE_STATUS = git_run(
    "status",
    "--porcelain"
)

# ------------------------------------------------------------
# 11. Evaluate acceptance criteria
# ------------------------------------------------------------

VERIFICATION_RESULTS = {
    "original_configuration_preserved": (
        sha256_file(ORIGINAL_CONFIG_PATH)
        == EXPECTED_ORIGINAL_HASH
    ),

    "environment_baseline_integrity": (
        ENVIRONMENT_BASELINE_CHECK
    ),

    "remote_repository_identity": (
        all(REPOSITORY_CHECKS.values())
    ),

    "remote_repository_empty": REMOTE_EMPTY_CHECK,

    "controlled_configuration_changes": (
        CONTROLLED_CHANGE_CHECK
    ),

    "revised_configuration_integrity": (
        REVISED_CONFIG_CHECK
    ),

    "configuration_change_record_integrity": (
        CHANGE_RECORD_CHECK
    ),

    "git_repository_initialized": (
        git_run("rev-parse", "--is-inside-work-tree")
        == "true"
    ),

    "local_branch_main": LOCAL_BRANCH == "main",

    "remote_configuration_correct": (
        CONFIGURED_REMOTE == REMOTE_URL
    ),

    "no_tracked_or_staged_files": (
        not bool(TRACKED_FILES)
    ),

    "clean_local_workspace": (
        not bool(WORKTREE_STATUS)
        and not bool(git_run("for-each-ref"))
    ),

    "clinical_deployment_restricted": (
        REVISED_CONFIG["clinical_deployment_authorised"]
        is False
    )
}

TOTAL_CHECKS = len(VERIFICATION_RESULTS)

PASSED_CHECKS = sum(
    VERIFICATION_RESULTS.values()
)

FAILED_CHECKS = TOTAL_CHECKS - PASSED_CHECKS

INITIALIZATION_STATUS = (
    "PASS" if FAILED_CHECKS == 0 else "FAIL"
)

# ------------------------------------------------------------
# 12. Persist Git initialization evidence
# ------------------------------------------------------------

RUN_TIMESTAMP = datetime.now(timezone.utc).strftime(
    "%Y%m%dT%H%M%S_%fZ"
)

VERIFICATION_REPORT = {
    "project": "HAT-DSP",
    "stage": "D00.3B",
    "verified_utc": datetime.now(timezone.utc).isoformat(),
    "execution_mode": EXECUTION_MODE,
    "git_version": GIT_VERSION,
    "git_workspace": str(GIT_WORKSPACE),
    "remote_repository": REPOSITORY_NAME,
    "remote_repository_id": REPOSITORY_METADATA["id"],
    "remote_url": REMOTE_URL,
    "repository_visibility": "public",
    "local_branch": LOCAL_BRANCH,
    "configuration_version": "1.0.1",
    "configuration_sha256": REVISED_CONFIG_HASH,
    "configuration_change_reference": "HAT-DSP-CR-001",
    "configuration_change_record_sha256": (
        sha256_file(CHANGE_REGISTER_PATH)
    ),
    "verification_results": VERIFICATION_RESULTS,
    "total_checks": TOTAL_CHECKS,
    "passed_checks": PASSED_CHECKS,
    "failed_checks": FAILED_CHECKS,
    "authenticated_write_access_verified": False,
    "initial_commit_completed": False,
    "remote_synchronization_verified": False,
    "initialization_status": INITIALIZATION_STATUS
}

REPORT_PATH = (
    LOG_DIR
    / f"D00_3B_git_initialization_{RUN_TIMESTAMP}.json"
)

write_json_exclusive(
    REPORT_PATH,
    VERIFICATION_REPORT
)

assert read_json(REPORT_PATH) == VERIFICATION_REPORT

# ------------------------------------------------------------
# 13. Display initialization results
# ------------------------------------------------------------

print("\nCONFIGURATION CHANGE SUMMARY")
print("-" * 75)

print("Change reference          : HAT-DSP-CR-001")
print("Previous configuration    : 1.0.0")
print("Revised configuration     : 1.0.1")
print("Original configuration    : PRESERVED")
print("Environment baseline      : PRESERVED")
print("Changed fields            :", len(CHANGED_FIELDS))

print("\nGIT REPOSITORY SUMMARY")
print("-" * 75)

print("Execution mode            :", EXECUTION_MODE)
print("Git version               :", GIT_VERSION)
print("Repository visibility     : PUBLIC")
print("Local branch              :", LOCAL_BRANCH)
print("Remote repository         :", REPOSITORY_NAME)
print("Remote access             : VERIFIED")
print("Authenticated write access: PENDING")
print("Initial commit            : PENDING")
print("Files staged              : 0")

print("\nFINAL VERIFICATION")
print("-" * 75)

for check, result in VERIFICATION_RESULTS.items():

    print(
        f"{check.replace('_', ' ').title():<48}: "
        f"{'PASS' if result else 'FAIL'}"
    )

print("\nACCEPTANCE SUMMARY")
print("-" * 75)

print("Total verification checks:", TOTAL_CHECKS)
print("Passed checks            :", PASSED_CHECKS)
print("Failed checks            :", FAILED_CHECKS)

print("\nLOCAL GIT WORKSPACE")
print(GIT_WORKSPACE)

print("\nCONFIGURED REMOTE")
print(CONFIGURED_REMOTE)

print("\nREVISED CONFIGURATION")
print(REVISED_CONFIG_PATH)

print("\nREVISED CONFIGURATION SHA-256")
print(REVISED_CONFIG_HASH)

print("\nCONFIGURATION CHANGE RECORD")
print(CHANGE_REGISTER_PATH)

print("\nVERIFICATION REPORT")
print(REPORT_PATH)

print("=" * 75)
print("D00.3B INITIALIZATION STATUS:", INITIALIZATION_STATUS)
print("=" * 75)

assert FAILED_CHECKS == 0, (
    "Git initialization acceptance criteria not satisfied."
)
```

Automated secret scanning shall not be interpreted as proof

```python

# ============================================================
# HAT-DSP
# D00.3D.1 — PUBLICATION SAFETY AND CONTENT VERIFICATION
# ============================================================

from pathlib import Path
from datetime import datetime, timezone

import hashlib
import subprocess
import json
import re
import csv

# ------------------------------------------------------------
# 1. Establish approved project artifact locations
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/"
    "HAT_DSP_MIVA_MIT_AI_CAPSTONE"
)

GOVERNANCE_DIR = PROJECT_ROOT / "00_project_governance"
LOG_DIR = PROJECT_ROOT / "23_logs"

GIT_WORKSPACE = Path(
    "/content/HAT_DSP_MIVA_MIT_AI_MASTER_CAPSTONE"
)

CONFIG_PATH = (
    GOVERNANCE_DIR / "project_config_v1_0_1.json"
)

STRUCTURE_REPORT_PATH = (
    LOG_DIR
    / "D00_3C_repository_structure_20261118T194118_856361Z.json"
)

SCAFFOLD_MANIFEST_PATH = (
    GOVERNANCE_DIR
    / "repository_scaffold_manifest_20261118T194118_856361Z.json"
)

EXPECTED_CONFIG_HASH = (
    "2e9d802885b8759caf0f17841d75fef0c"
    "de572cefe4f150846177cd0f19ea4ad"
)

def sha256_file(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()

def read_json(path):
    return json.loads(path.read_text(encoding="utf-8"))

def git_run(*arguments, check=True):
    return subprocess.run(
        ["git", "-C", str(GIT_WORKSPACE), *arguments],
        capture_output=True,
        text=True,
        check=check
    )

print("=" * 75)
print("HAT-DSP | D00.3D.1")
print("PUBLICATION SAFETY AND CONTENT VERIFICATION")
print("=" * 75)

# ------------------------------------------------------------
# 2. Verify prerequisite evidence
# ------------------------------------------------------------

for path in [
    CONFIG_PATH,
    STRUCTURE_REPORT_PATH,
    SCAFFOLD_MANIFEST_PATH
]:
    assert path.is_file(), f"Required artifact missing: {path}"

assert (GIT_WORKSPACE / ".git").is_dir()

CONFIG_INTEGRITY = (
    sha256_file(CONFIG_PATH) == EXPECTED_CONFIG_HASH
)

assert CONFIG_INTEGRITY

PROJECT_CONFIG = read_json(CONFIG_PATH)

STRUCTURE_REPORT = read_json(
    STRUCTURE_REPORT_PATH
)

SCAFFOLD_MANIFEST = read_json(
    SCAFFOLD_MANIFEST_PATH
)

STRUCTURE_ACCEPTED = (
    STRUCTURE_REPORT["structure_status"] == "PASS"
    and STRUCTURE_REPORT["failed_checks"] == 0
)

MANIFEST_INTEGRITY = (
    sha256_file(SCAFFOLD_MANIFEST_PATH)
    == STRUCTURE_REPORT["scaffold_manifest_sha256"]
)

assert STRUCTURE_ACCEPTED
assert MANIFEST_INTEGRITY

assert (
    SCAFFOLD_MANIFEST["configuration_sha256"]
    == EXPECTED_CONFIG_HASH
)

assert (
    PROJECT_CONFIG["github_repository_visibility"]
    == "public"
)

REMOTE_INTEGRITY = (
    git_run(
        "remote", "get-url", "origin"
    ).stdout.strip()
    == PROJECT_CONFIG["github_remote_url"]
)

BRANCH_INTEGRITY = (
    git_run(
        "symbolic-ref", "--short", "HEAD"
    ).stdout.strip() == "main"
)

assert REMOTE_INTEGRITY
assert BRANCH_INTEGRITY

# ------------------------------------------------------------
# 3. Verify repository candidate allowlist
# ------------------------------------------------------------

APPROVED_FILES = sorted(
    SCAFFOLD_MANIFEST["scaffold_files"].keys()
)

assert len(APPROVED_FILES) == 19

ACTUAL_CANDIDATES = sorted(
    git_run(
        "ls-files",
        "--others",
        "--exclude-standard"
    ).stdout.splitlines()
)

CANDIDATE_INTEGRITY = (
    ACTUAL_CANDIDATES == APPROVED_FILES
    and APPROVED_FILES
    == sorted(STRUCTURE_REPORT["expected_candidates"])
)

assert CANDIDATE_INTEGRITY, (
    "Unexpected or missing repository candidates."
)

assert not git_run("ls-files").stdout.strip()
assert not git_run("for-each-ref").stdout.strip()

# ------------------------------------------------------------
# 4. Define preliminary secret scanning rules
# ------------------------------------------------------------

SECRET_PATTERNS = {
    "github_personal_access_token": (
        r"\bgh[pousr]_[A-Za-z0-9]{20,}\b"
    ),

    "github_fine_grained_token": (
        r"\bgithub_pat_[A-Za-z0-9_]{20,}\b"
    ),

    "aws_access_key": (
        r"\b(?:AKIA|ASIA)[A-Z0-9]{16}\b"
    ),

    "google_api_key": (
        r"\bAIza[A-Za-z0-9_-]{35}\b"
    ),

    "openai_style_api_key": (
        r"\bsk-[A-Za-z0-9_-]{20,}\b"
    ),

    "private_key_material": (
        r"-----BEGIN [A-Z0-9 ]*PRIVATE KEY-----"
    ),

    "embedded_url_credentials": (
        r"https?://[^/\s:@]+:[^@\s]+@"
    ),

    "hardcoded_secret_assignment": (
        r"""(?im)^\s*(?:api[_-]?key|secret|token|"""
        r"""password|access[_-]?token|client[_-]?secret)"""
        r"""\s*[:=]\s*["'][^"'\r\n]{8,}["']"""
    )
}

COMPILED_PATTERNS = {
    name: re.compile(pattern)
    for name, pattern in SECRET_PATTERNS.items()
}

# ------------------------------------------------------------
# 5. Verify file integrity and scan content
# ------------------------------------------------------------

REVIEW_RECORDS = []
SECRET_FINDINGS = []

print("\nPUBLICATION CANDIDATE VERIFICATION")
print("-" * 75)

for relative_path in APPROVED_FILES:

    path = GIT_WORKSPACE / relative_path

    reference = SCAFFOLD_MANIFEST[
        "scaffold_files"
    ][relative_path]

    safe_path = (
        path.is_file()
        and not path.is_symlink()
        and path.resolve().is_relative_to(
            GIT_WORKSPACE.resolve()
        )
    )

    assert safe_path, (
        f"Unsafe or missing candidate: {relative_path}"
    )

    current_hash = sha256_file(path)

    hash_matches = (
        current_hash == reference["sha256"]
    )

    size_matches = (
        path.stat().st_size
        == reference["size_bytes"]
    )

    content = path.read_text(encoding="utf-8")

    file_findings = []

    for rule_name, pattern in COMPILED_PATTERNS.items():

        for match in pattern.finditer(content):

            finding = {
                "file": relative_path,
                "rule": rule_name,
                "line_number": (
                    content.count(
                        "\n", 0, match.start()
                    ) + 1
                )
            }

            # Do not record or display matched secret values.
            file_findings.append(finding)
            SECRET_FINDINGS.append(finding)

    placeholder_valid = (
        path.name != ".gitkeep"
        or path.stat().st_size == 0
    )

    REVIEW_RECORDS.append({
        "file": relative_path,
        "size_bytes": path.stat().st_size,
        "sha256": current_hash,
        "hash_integrity": hash_matches,
        "size_integrity": size_matches,
        "safe_file_location": safe_path,
        "placeholder_valid": placeholder_valid,
        "secret_finding_count": len(file_findings),
        "manual_review_status": (
            "PENDING"
            if path.name != ".gitkeep"
            else "EMPTY_PLACEHOLDER_VERIFIED"
        )
    })

    file_passed = (
        hash_matches
        and size_matches
        and safe_path
        and placeholder_valid
        and not file_findings
    )

    print(
        f"{relative_path:<45}: "
        f"{'PASS' if file_passed else 'FAIL'}"
    )

# ------------------------------------------------------------
# 6. Verify requirements file publication safety
# ------------------------------------------------------------

REQUIREMENTS_PATH = (
    GIT_WORKSPACE / "requirements.txt"
)

REQUIREMENT_LINES = [
    line.strip()
    for line in REQUIREMENTS_PATH.read_text(
        encoding="utf-8"
    ).splitlines()
    if line.strip()
]

PINNED_REQUIREMENT_PATTERN = re.compile(
    r"[A-Za-z0-9_.-]+==[A-Za-z0-9_.+!-]+"
)

REQUIREMENTS_FORMAT_CHECK = (
    len(REQUIREMENT_LINES) == 16
    and all(
        PINNED_REQUIREMENT_PATTERN.fullmatch(line)
        is not None
        for line in REQUIREMENT_LINES
    )
)

# ------------------------------------------------------------
# 7. Verify complete repository safety assessment
# ------------------------------------------------------------

VERIFICATION_RESULTS = {
    "project_configuration_integrity": CONFIG_INTEGRITY,

    "repository_structure_accepted": STRUCTURE_ACCEPTED,

    "scaffold_manifest_integrity": MANIFEST_INTEGRITY,

    "remote_configuration_integrity": REMOTE_INTEGRITY,

    "local_branch_integrity": BRANCH_INTEGRITY,

    "publication_candidate_allowlist": CANDIDATE_INTEGRITY,

    "scaffold_file_hashes_preserved": all(
        record["hash_integrity"]
        for record in REVIEW_RECORDS
    ),

    "scaffold_file_sizes_preserved": all(
        record["size_integrity"]
        for record in REVIEW_RECORDS
    ),

    "safe_candidate_file_locations": all(
        record["safe_file_location"]
        for record in REVIEW_RECORDS
    ),

    "empty_placeholders_verified": all(
        record["placeholder_valid"]
        for record in REVIEW_RECORDS
    ),

    "preliminary_secret_scan": (
        len(SECRET_FINDINGS) == 0
    ),

    "requirements_publication_format": (
        REQUIREMENTS_FORMAT_CHECK
    ),

    "no_files_staged_or_tracked": (
        not bool(git_run("ls-files").stdout.strip())
    ),

    "no_commits_created": (
        not bool(git_run("for-each-ref").stdout.strip())
    ),

    "clinical_deployment_restricted": (
        PROJECT_CONFIG["clinical_deployment_authorised"]
        is False
    )
}

TOTAL_CHECKS = len(VERIFICATION_RESULTS)

PASSED_CHECKS = sum(
    VERIFICATION_RESULTS.values()
)

FAILED_CHECKS = TOTAL_CHECKS - PASSED_CHECKS

ASSESSMENT_STATUS = (
    "PASS" if FAILED_CHECKS == 0 else "FAIL"
)

# ------------------------------------------------------------
# 8. Persist publication review evidence
# ------------------------------------------------------------

RUN_TIMESTAMP = datetime.now(timezone.utc).strftime(
    "%Y%m%dT%H%M%S_%fZ"
)

REVIEW_REGISTER_PATH = (
    GOVERNANCE_DIR
    / f"publication_review_register_{RUN_TIMESTAMP}.csv"
)

with REVIEW_REGISTER_PATH.open(
    "x", newline="", encoding="utf-8"
) as file:

    writer = csv.DictWriter(
        file,
        fieldnames=list(REVIEW_RECORDS[0].keys())
    )

    writer.writeheader()
    writer.writerows(REVIEW_RECORDS)

VERIFICATION_REPORT = {
    "project": "HAT-DSP",
    "stage": "D00.3D.1",
    "verified_utc": datetime.now(timezone.utc).isoformat(),
    "configuration_version": "1.0.1",
    "configuration_sha256": EXPECTED_CONFIG_HASH,
    "repository_visibility": "public",
    "scaffold_manifest_sha256": (
        sha256_file(SCAFFOLD_MANIFEST_PATH)
    ),
    "publication_candidates": APPROVED_FILES,
    "files_reviewed_automatically": len(REVIEW_RECORDS),
    "secret_scanning_method": (
        "Preliminary regular-expression content scanning"
    ),
    "secret_scanning_rules": len(SECRET_PATTERNS),
    "secret_finding_count": len(SECRET_FINDINGS),
    "secret_findings": SECRET_FINDINGS,
    "file_review_records": REVIEW_RECORDS,
    "review_register": REVIEW_REGISTER_PATH.name,
    "review_register_sha256": (
        sha256_file(REVIEW_REGISTER_PATH)
    ),
    "verification_results": VERIFICATION_RESULTS,
    "total_checks": TOTAL_CHECKS,
    "passed_checks": PASSED_CHECKS,
    "failed_checks": FAILED_CHECKS,
    "manual_publication_review_completed": False,
    "publication_authorised": False,
    "initial_commit_completed": False,
    "remote_push_completed": False,
    "assessment_status": ASSESSMENT_STATUS
}

REPORT_PATH = (
    LOG_DIR
    / f"D00_3D_1_publication_safety_{RUN_TIMESTAMP}.json"
)

with REPORT_PATH.open("x", encoding="utf-8") as file:
    json.dump(VERIFICATION_REPORT, file, indent=4)

assert read_json(REPORT_PATH) == VERIFICATION_REPORT

# ------------------------------------------------------------
# 9. Display assessment results
# ------------------------------------------------------------

print("\nPUBLICATION SAFETY VERIFICATION")
print("-" * 75)

for check, result in VERIFICATION_RESULTS.items():

    print(
        f"{check.replace('_', ' ').title():<48}: "
        f"{'PASS' if result else 'FAIL'}"
    )

print("\nPUBLICATION ASSESSMENT SUMMARY")
print("-" * 75)

print("Publication candidates       :", len(APPROVED_FILES))
print("Files scanned                :", len(REVIEW_RECORDS))
print("Secret scanning rules        :", len(SECRET_PATTERNS))
print("Potential secret findings    :", len(SECRET_FINDINGS))
print("Pinned direct dependencies   :", len(REQUIREMENT_LINES))
print("Total verification checks    :", TOTAL_CHECKS)
print("Passed checks                :", PASSED_CHECKS)
print("Failed checks                :", FAILED_CHECKS)
print("Manual publication review    : PENDING")
print("Publication authorisation    : PENDING")
print("Initial commit               : NOT PERFORMED")
print("Remote push                  : NOT PERFORMED")

if SECRET_FINDINGS:

    print("\nPOTENTIAL SECRET FINDING LOCATIONS")
    print(json.dumps(SECRET_FINDINGS, indent=4))

print("\nPUBLICATION REVIEW REGISTER")
print(REVIEW_REGISTER_PATH)

print("\nVERIFICATION REPORT")
print(REPORT_PATH)

print("=" * 75)
print("D00.3D.1 AUTOMATED ASSESSMENT:", ASSESSMENT_STATUS)
print("=" * 75)

assert FAILED_CHECKS == 0, (
    "Publication safety assessment failed."
)

# ------------------------------------------------------------
# 10. Present actual file contents for manual review
# ------------------------------------------------------------

print("\nMANUAL PUBLICATION REVIEW")
print("=" * 75)

for filename in [
    "README.md",
    "SECURITY.md",
    "requirements.txt",
    ".gitignore"
]:

    print(f"\nFILE: {filename}")
    print("-" * 75)

    print(
        (GIT_WORKSPACE / filename).read_text(
            encoding="utf-8"
        )
    )

print("=" * 75)
print("AWAITING MANUAL REVIEW AND COMMIT AUTHORISATION")
print("=" * 75)
```

```python

# ============================================================
# HAT-DSP
# D00.3D.1B — REVISED PUBLICATION SAFETY ASSESSMENT
# ============================================================

from pathlib import Path
from datetime import datetime, timezone

import hashlib
import subprocess
import json
import re
import csv

# ------------------------------------------------------------
# 1. Establish approved artifact locations
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/"
    "HAT_DSP_MIVA_MIT_AI_CAPSTONE"
)

GOVERNANCE_DIR = PROJECT_ROOT / "00_project_governance"
LOG_DIR = PROJECT_ROOT / "23_logs"

GIT_WORKSPACE = Path(
    "/content/HAT_DSP_MIVA_MIT_AI_MASTER_CAPSTONE"
)

CONFIG_PATH = (
    GOVERNANCE_DIR / "project_config_v1_0_1.json"
)

REVISION_REPORT_PATH = (
    LOG_DIR
    / "D00_3D_1A_scaffold_revision_20261118T195003_139428Z.json"
)

REVISED_MANIFEST_PATH = (
    GOVERNANCE_DIR
    / "repository_scaffold_manifest_v1_0_1_20261118T195003_139428Z.json"
)

ORIGINAL_MANIFEST_PATH = (
    GOVERNANCE_DIR
    / "repository_scaffold_manifest_20261118T194118_856361Z.json"
)

ORIGINAL_ASSESSMENT_PATH = (
    LOG_DIR
    / "D00_3D_1_publication_safety_20261118T194526_170421Z.json"
)

EXPECTED_CONFIG_HASH = (
    "2e9d802885b8759caf0f17841d75fef0c"
    "de572cefe4f150846177cd0f19ea4ad"
)

NOTEBOOK_PATH = (
    "notebooks/"
    "MIVA_MIT_AI_HAT_DSP_MASTER'S_CAPSTONE_PROJECT.ipynb"
)

# ------------------------------------------------------------
# 2. Define reusable verification functions
# ------------------------------------------------------------

def sha256_file(path):
    return hashlib.sha256(
        path.read_bytes()
    ).hexdigest()


def read_json(path):
    return json.loads(
        path.read_text(encoding="utf-8")
    )


def git_run(*arguments, check=True):

    return subprocess.run(
        ["git", "-C", str(GIT_WORKSPACE), *arguments],
        capture_output=True,
        text=True,
        check=check
    )


def is_ignored(relative_path):

    result = git_run(
        "check-ignore",
        "--no-index",
        "--",
        relative_path,
        check=False
    )

    assert result.returncode in (0, 1)

    return result.returncode == 0


print("=" * 75)
print("HAT-DSP | D00.3D.1B")
print("REVISED PUBLICATION SAFETY ASSESSMENT")
print("=" * 75)

# ------------------------------------------------------------
# 3. Verify prerequisite evidence
# ------------------------------------------------------------

REQUIRED_PATHS = [
    CONFIG_PATH,
    REVISION_REPORT_PATH,
    REVISED_MANIFEST_PATH,
    ORIGINAL_MANIFEST_PATH,
    ORIGINAL_ASSESSMENT_PATH
]

for path in REQUIRED_PATHS:
    assert path.is_file(), f"Missing artifact: {path}"

assert (GIT_WORKSPACE / ".git").is_dir()

CONFIG_INTEGRITY = (
    sha256_file(CONFIG_PATH) == EXPECTED_CONFIG_HASH
)

assert CONFIG_INTEGRITY

PROJECT_CONFIG = read_json(CONFIG_PATH)
REVISION_REPORT = read_json(REVISION_REPORT_PATH)
REVISED_MANIFEST = read_json(REVISED_MANIFEST_PATH)
ORIGINAL_MANIFEST = read_json(ORIGINAL_MANIFEST_PATH)

REVISION_ACCEPTED = (
    REVISION_REPORT["revision_status"] == "PASS"
    and REVISION_REPORT["failed_checks"] == 0
    and REVISION_REPORT["change_reference"]
    == "HAT-DSP-SCR-001"
)

MANIFEST_INTEGRITY = (
    sha256_file(REVISED_MANIFEST_PATH)
    == REVISION_REPORT["revised_manifest_sha256"]
)

ORIGINAL_EVIDENCE_INTEGRITY = (
    sha256_file(ORIGINAL_MANIFEST_PATH)
    == REVISION_REPORT["original_manifest_sha256"]
    and sha256_file(ORIGINAL_ASSESSMENT_PATH)
    == REVISION_REPORT["original_assessment_sha256"]
)

CONFIGURATION_LINKAGE = (
    REVISED_MANIFEST["configuration_sha256"]
    == EXPECTED_CONFIG_HASH
    and REVISION_REPORT["configuration_sha256"]
    == EXPECTED_CONFIG_HASH
    and REVISED_MANIFEST["scaffold_version"]
    == "1.0.1"
    and REVISED_MANIFEST["official_notebook_path"]
    == NOTEBOOK_PATH
)

assert REVISION_ACCEPTED
assert MANIFEST_INTEGRITY
assert ORIGINAL_EVIDENCE_INTEGRITY
assert CONFIGURATION_LINKAGE

print("\nPREREQUISITE VERIFICATION")
print("-" * 75)

print("Project configuration integrity : PASS")
print("Scaffold revision acceptance    : PASS")
print("Revised manifest integrity      : PASS")
print("Original evidence preservation  : PASS")
print("Configuration linkage           : PASS")

# ------------------------------------------------------------
# 4. Verify repository identity and publication candidates
# ------------------------------------------------------------

REMOTE_INTEGRITY = (
    git_run(
        "remote", "get-url", "origin"
    ).stdout.strip()
    == PROJECT_CONFIG["github_remote_url"]
)

BRANCH_INTEGRITY = (
    git_run(
        "symbolic-ref", "--short", "HEAD"
    ).stdout.strip() == "main"
)

APPROVED_FILES = sorted(
    REVISED_MANIFEST["scaffold_files"]
)

assert len(APPROVED_FILES) == 20

ACTUAL_CANDIDATES = sorted(
    git_run(
        "ls-files",
        "--others",
        "--exclude-standard"
    ).stdout.splitlines()
)

CANDIDATE_INTEGRITY = (
    ACTUAL_CANDIDATES == APPROVED_FILES
    and APPROVED_FILES
    == sorted(REVISION_REPORT["expected_candidates"])
)

assert REMOTE_INTEGRITY
assert BRANCH_INTEGRITY
assert CANDIDATE_INTEGRITY

assert not git_run("ls-files").stdout.strip()
assert not git_run("for-each-ref").stdout.strip()

# ------------------------------------------------------------
# 5. Verify controlled scaffold change scope
# ------------------------------------------------------------

ORIGINAL_FILES = ORIGINAL_MANIFEST["scaffold_files"]
REVISED_FILES = REVISED_MANIFEST["scaffold_files"]

EXPECTED_MODIFICATIONS = {
    ".gitignore",
    "README.md",
    "SECURITY.md"
}

EXPECTED_ADDITIONS = {
    "notebooks/.gitkeep"
}

ACTUAL_MODIFICATIONS = {
    name
    for name in ORIGINAL_FILES
    if name in REVISED_FILES
    and ORIGINAL_FILES[name]["sha256"]
    != REVISED_FILES[name]["sha256"]
}

ACTUAL_ADDITIONS = (
    set(REVISED_FILES) - set(ORIGINAL_FILES)
)

ACTUAL_REMOVALS = (
    set(ORIGINAL_FILES) - set(REVISED_FILES)
)

CHANGE_SCOPE_INTEGRITY = (
    ACTUAL_MODIFICATIONS == EXPECTED_MODIFICATIONS
    and ACTUAL_ADDITIONS == EXPECTED_ADDITIONS
    and not ACTUAL_REMOVALS
)

# ------------------------------------------------------------
# 6. Define original preliminary secret scanning rules
# ------------------------------------------------------------

SECRET_PATTERNS = {
    "github_personal_access_token": (
        r"\bgh[pousr]_[A-Za-z0-9]{20,}\b"
    ),

    "github_fine_grained_token": (
        r"\bgithub_pat_[A-Za-z0-9_]{20,}\b"
    ),

    "aws_access_key": (
        r"\b(?:AKIA|ASIA)[A-Z0-9]{16}\b"
    ),

    "google_api_key": (
        r"\bAIza[A-Za-z0-9_-]{35}\b"
    ),

    "openai_style_api_key": (
        r"\bsk-[A-Za-z0-9_-]{20,}\b"
    ),

    "private_key_material": (
        r"-----BEGIN [A-Z0-9 ]*PRIVATE KEY-----"
    ),

    "embedded_url_credentials": (
        r"https?://[^/\s:@]+:[^@\s]+@"
    ),

    "hardcoded_secret_assignment": (
        r"""(?im)^\s*(?:api[_-]?key|secret|token|"""
        r"""password|access[_-]?token|client[_-]?secret)"""
        r"""\s*[:=]\s*["'][^"'\r\n]{8,}["']"""
    )
}

COMPILED_PATTERNS = {
    name: re.compile(pattern)
    for name, pattern in SECRET_PATTERNS.items()
}

# ------------------------------------------------------------
# 7. Reassess every revised publication candidate
# ------------------------------------------------------------

REVIEW_RECORDS = []
SECRET_FINDINGS = []

print("\nREVISED PUBLICATION CANDIDATES")
print("-" * 75)

for relative_path in APPROVED_FILES:

    path = GIT_WORKSPACE / relative_path
    reference = REVISED_FILES[relative_path]

    SAFE_PATH = (
        path.is_file()
        and not path.is_symlink()
        and path.resolve().is_relative_to(
            GIT_WORKSPACE.resolve()
        )
    )

    assert SAFE_PATH, (
        f"Unsafe candidate path: {relative_path}"
    )

    CURRENT_HASH = sha256_file(path)

    HASH_CHECK = (
        CURRENT_HASH == reference["sha256"]
    )

    SIZE_CHECK = (
        path.stat().st_size
        == reference["size_bytes"]
    )

    CONTENT = path.read_text(encoding="utf-8")

    FILE_FINDINGS = []

    for rule_name, pattern in COMPILED_PATTERNS.items():

        for match in pattern.finditer(CONTENT):

            finding = {
                "file": relative_path,
                "rule": rule_name,
                "line_number": (
                    CONTENT.count(
                        "\n", 0, match.start()
                    ) + 1
                )
            }

            FILE_FINDINGS.append(finding)
            SECRET_FINDINGS.append(finding)

    PLACEHOLDER_CHECK = (
        path.name != ".gitkeep"
        or path.stat().st_size == 0
    )

    if relative_path in EXPECTED_ADDITIONS:
        change_status = "ADDED"

    elif relative_path in EXPECTED_MODIFICATIONS:
        change_status = "MODIFIED"

    else:
        change_status = "PRESERVED"

    FILE_PASSED = (
        HASH_CHECK
        and SIZE_CHECK
        and SAFE_PATH
        and PLACEHOLDER_CHECK
        and not FILE_FINDINGS
    )

    REVIEW_RECORDS.append({
        "file": relative_path,
        "change_status": change_status,
        "size_bytes": path.stat().st_size,
        "sha256": CURRENT_HASH,
        "hash_integrity": HASH_CHECK,
        "size_integrity": SIZE_CHECK,
        "safe_file_location": SAFE_PATH,
        "placeholder_valid": PLACEHOLDER_CHECK,
        "secret_finding_count": len(FILE_FINDINGS),
        "automated_assessment_passed": FILE_PASSED,
        "manual_review_status": (
            "EMPTY_PLACEHOLDER_VERIFIED"
            if path.name == ".gitkeep"
            else "PENDING"
        )
    })

    print(
        f"{relative_path:<35}"
        f"{change_status:<12}: "
        f"{'PASS' if FILE_PASSED else 'FAIL'}"
    )

# ------------------------------------------------------------
# 8. Reverify notebook publication controls
# ------------------------------------------------------------

NOTEBOOK_POLICY_TESTS = {
    "official_master_notebook": (
        not is_ignored(NOTEBOOK_PATH)
    ),

    "notebook_placeholder": (
        not is_ignored("notebooks/.gitkeep")
    ),

    "other_notebook_excluded": (
        is_ignored("notebooks/experimental.ipynb")
    ),

    "root_notebook_excluded": (
        is_ignored("temporary.ipynb")
    ),

    "source_notebook_excluded": (
        is_ignored("src/analysis.ipynb")
    ),

    "checkpoint_excluded": (
        is_ignored(
            "notebooks/.ipynb_checkpoints/checkpoint.ipynb"
        )
    ),

    "notebook_credentials_excluded": (
        is_ignored("notebooks/credentials.json")
    ),

    "notebook_dataset_excluded": (
        is_ignored("notebooks/patient_records.csv")
    ),

    "notebook_log_excluded": (
        is_ignored("notebooks/execution.log")
    ),

    "notebook_model_excluded": (
        is_ignored("notebooks/model.joblib")
    )
}

NOTEBOOK_POLICY_CHECK = all(
    NOTEBOOK_POLICY_TESTS.values()
)

NOTEBOOK_ABSENT_CHECK = (
    not (GIT_WORKSPACE / NOTEBOOK_PATH).exists()
)

# ------------------------------------------------------------
# 9. Verify dependency specification
# ------------------------------------------------------------

REQUIREMENTS_PATH = (
    GIT_WORKSPACE / "requirements.txt"
)

REQUIREMENT_LINES = [
    line.strip()
    for line in REQUIREMENTS_PATH.read_text(
        encoding="utf-8"
    ).splitlines()
    if line.strip()
]

REQUIREMENTS_CHECK = (
    len(REQUIREMENT_LINES) == 16
    and all(
        re.fullmatch(
            r"[A-Za-z0-9_.-]+==[A-Za-z0-9_.+!-]+",
            line
        )
        for line in REQUIREMENT_LINES
    )
    and sha256_file(REQUIREMENTS_PATH)
    == ORIGINAL_FILES["requirements.txt"]["sha256"]
)

# ------------------------------------------------------------
# 10. Evaluate acceptance criteria
# ------------------------------------------------------------

VERIFICATION_RESULTS = {
    "project_configuration_integrity": CONFIG_INTEGRITY,

    "scaffold_revision_accepted": REVISION_ACCEPTED,

    "revised_manifest_integrity": MANIFEST_INTEGRITY,

    "original_evidence_preserved": (
        ORIGINAL_EVIDENCE_INTEGRITY
    ),

    "configuration_linkage_verified": (
        CONFIGURATION_LINKAGE
    ),

    "controlled_change_scope": CHANGE_SCOPE_INTEGRITY,

    "remote_configuration_integrity": REMOTE_INTEGRITY,

    "local_branch_integrity": BRANCH_INTEGRITY,

    "revised_candidate_allowlist": CANDIDATE_INTEGRITY,

    "all_scaffold_file_hashes_verified": all(
        record["hash_integrity"]
        for record in REVIEW_RECORDS
    ),

    "all_scaffold_file_sizes_verified": all(
        record["size_integrity"]
        for record in REVIEW_RECORDS
    ),

    "safe_candidate_file_locations": all(
        record["safe_file_location"]
        for record in REVIEW_RECORDS
    ),

    "empty_placeholders_verified": all(
        record["placeholder_valid"]
        for record in REVIEW_RECORDS
    ),

    "preliminary_secret_scan": (
        len(SECRET_FINDINGS) == 0
    ),

    "approved_requirements_preserved": (
        REQUIREMENTS_CHECK
    ),

    "notebook_publication_controls": (
        NOTEBOOK_POLICY_CHECK
    ),

    "no_premature_notebook_import": (
        NOTEBOOK_ABSENT_CHECK
    ),

    "no_files_staged_or_tracked": (
        not bool(git_run("ls-files").stdout.strip())
    ),

    "no_commits_created": (
        not bool(git_run("for-each-ref").stdout.strip())
    ),

    "clinical_deployment_restricted": (
        PROJECT_CONFIG["clinical_deployment_authorised"]
        is False
    )
}

TOTAL_CHECKS = len(VERIFICATION_RESULTS)
PASSED_CHECKS = sum(VERIFICATION_RESULTS.values())
FAILED_CHECKS = TOTAL_CHECKS - PASSED_CHECKS

ASSESSMENT_STATUS = (
    "PASS" if FAILED_CHECKS == 0 else "FAIL"
)

# ------------------------------------------------------------
# 11. Persist reassessment evidence
# ------------------------------------------------------------

RUN_TIMESTAMP = datetime.now(timezone.utc).strftime(
    "%Y%m%dT%H%M%S_%fZ"
)

REGISTER_PATH = (
    GOVERNANCE_DIR
    / f"revised_publication_review_register_{RUN_TIMESTAMP}.csv"
)

with REGISTER_PATH.open(
    "x", newline="", encoding="utf-8"
) as file:

    writer = csv.DictWriter(
        file,
        fieldnames=list(REVIEW_RECORDS[0].keys())
    )

    writer.writeheader()
    writer.writerows(REVIEW_RECORDS)

with REGISTER_PATH.open(
    newline="", encoding="utf-8"
) as file:

    assert len(list(csv.DictReader(file))) == 20

ASSESSMENT_REPORT = {
    "project": "HAT-DSP",
    "stage": "D00.3D.1B",
    "verified_utc": datetime.now(timezone.utc).isoformat(),
    "configuration_version": "1.0.1",
    "scaffold_version": "1.0.1",
    "configuration_sha256": EXPECTED_CONFIG_HASH,
    "scaffold_change_reference": "HAT-DSP-SCR-001",
    "repository_visibility": "public",
    "scaffold_manifest_sha256": (
        sha256_file(REVISED_MANIFEST_PATH)
    ),
    "scaffold_revision_report_sha256": (
        sha256_file(REVISION_REPORT_PATH)
    ),
    "publication_candidates": APPROVED_FILES,
    "files_reviewed_automatically": len(REVIEW_RECORDS),
    "modified_files": sorted(ACTUAL_MODIFICATIONS),
    "added_files": sorted(ACTUAL_ADDITIONS),
    "removed_files": sorted(ACTUAL_REMOVALS),
    "secret_scanning_method": (
        "Preliminary regular-expression content scanning"
    ),
    "secret_scanning_rules": len(SECRET_PATTERNS),
    "secret_finding_count": len(SECRET_FINDINGS),
    "secret_findings": SECRET_FINDINGS,
    "notebook_policy_tests": NOTEBOOK_POLICY_TESTS,
    "file_review_records": REVIEW_RECORDS,
    "review_register": REGISTER_PATH.name,
    "review_register_sha256": sha256_file(REGISTER_PATH),
    "verification_results": VERIFICATION_RESULTS,
    "total_checks": TOTAL_CHECKS,
    "passed_checks": PASSED_CHECKS,
    "failed_checks": FAILED_CHECKS,
    "manual_publication_review_completed": False,
    "publication_authorised": False,
    "notebook_import_completed": False,
    "initial_commit_completed": False,
    "remote_push_completed": False,
    "assessment_status": ASSESSMENT_STATUS
}

REPORT_PATH = (
    LOG_DIR
    / f"D00_3D_1B_revised_publication_safety_{RUN_TIMESTAMP}.json"
)

with REPORT_PATH.open("x", encoding="utf-8") as file:

    json.dump(
        ASSESSMENT_REPORT,
        file,
        indent=4
    )

assert read_json(REPORT_PATH) == ASSESSMENT_REPORT

# ------------------------------------------------------------
# 12. Display verification results
# ------------------------------------------------------------

print("\nFINAL PUBLICATION SAFETY VERIFICATION")
print("-" * 75)

for check, result in VERIFICATION_RESULTS.items():

    print(
        f"{check.replace('_', ' ').title():<48}: "
        f"{'PASS' if result else 'FAIL'}"
    )

print("\nREVISED PUBLICATION ASSESSMENT SUMMARY")
print("-" * 75)

print("Scaffold version             : 1.0.1")
print("Publication candidates       :", len(APPROVED_FILES))
print("Files scanned                :", len(REVIEW_RECORDS))
print("Modified files               :", len(ACTUAL_MODIFICATIONS))
print("Added files                  :", len(ACTUAL_ADDITIONS))
print("Removed files                :", len(ACTUAL_REMOVALS))
print("Secret scanning rules        :", len(SECRET_PATTERNS))
print("Potential secret findings    :", len(SECRET_FINDINGS))
print("Pinned direct dependencies   :", len(REQUIREMENT_LINES))

print(
    "Notebook policy tests        :",
    f"{sum(NOTEBOOK_POLICY_TESTS.values())}/"
    f"{len(NOTEBOOK_POLICY_TESTS)}"
)

print("Total verification checks    :", TOTAL_CHECKS)
print("Passed checks                :", PASSED_CHECKS)
print("Failed checks                :", FAILED_CHECKS)
print("Manual publication review    : PENDING")
print("Publication authorisation    : PENDING")
print("Official notebook imported   : NOT YET")
print("Initial commit               : NOT PERFORMED")
print("Remote push                  : NOT PERFORMED")
print("Verification report integrity: PASS")

if SECRET_FINDINGS:

    print("\nPOTENTIAL SECRET FINDING LOCATIONS")
    print(json.dumps(SECRET_FINDINGS, indent=4))

print("\nREVISED PUBLICATION REVIEW REGISTER")
print(REGISTER_PATH)

print("\nVERIFICATION REPORT")
print(REPORT_PATH)

print("=" * 75)
print("D00.3D.1B REVISED ASSESSMENT STATUS:", ASSESSMENT_STATUS)
print("=" * 75)

assert FAILED_CHECKS == 0, (
    "Revised publication safety assessment failed."
)

# ------------------------------------------------------------
# 13. Present only newly modified content for manual review
# ------------------------------------------------------------

import difflib

BACKUP_DIR = (
    GOVERNANCE_DIR
    / "repository_scaffold_history"
    / "v1.0.0"
)

print("\nMANUAL REVIEW — CONTROLLED DOCUMENT CHANGES")
print("=" * 75)

for filename in sorted(EXPECTED_MODIFICATIONS):

    original_path = BACKUP_DIR / filename
    revised_path = GIT_WORKSPACE / filename

    assert (
        sha256_file(original_path)
        == ORIGINAL_FILES[filename]["sha256"]
    )

    ORIGINAL_LINES = original_path.read_text(
        encoding="utf-8"
    ).splitlines()

    REVISED_LINES = revised_path.read_text(
        encoding="utf-8"
    ).splitlines()

    print(f"\nFILE: {filename}")
    print("-" * 75)

    CHANGES = difflib.unified_diff(
        ORIGINAL_LINES,
        REVISED_LINES,
        fromfile=f"v1.0.0/{filename}",
        tofile=f"v1.0.1/{filename}",
        lineterm=""
    )

    print("\n".join(CHANGES))

print("=" * 75)
print("AWAITING MANUAL REVIEW AND COMMIT AUTHORISATION")
print("=" * 75)
```

```python

# ============================================================
# HAT-DSP
# D00.3D.2 — INITIAL CONTROLLED GIT COMMIT
# ============================================================

from pathlib import Path
from datetime import datetime, timezone

import hashlib
import subprocess
import json
import re

# ------------------------------------------------------------
# 1. Establish approved artifact locations
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/"
    "HAT_DSP_MIVA_MIT_AI_CAPSTONE"
)

GOVERNANCE_DIR = PROJECT_ROOT / "00_project_governance"
LOG_DIR = PROJECT_ROOT / "23_logs"

GIT_WORKSPACE = Path(
    "/content/HAT_DSP_MIVA_MIT_AI_MASTER_CAPSTONE"
)

CONFIG_PATH = (
    GOVERNANCE_DIR / "project_config_v1_0_1.json"
)

MANIFEST_PATH = (
    GOVERNANCE_DIR
    / "repository_scaffold_manifest_v1_0_1_20261118T195003_139428Z.json"
)

ASSESSMENT_PATH = (
    LOG_DIR
    / "D00_3D_1B_revised_publication_safety_20261118T195253_355357Z.json"
)

EXPECTED_CONFIG_HASH = (
    "2e9d802885b8759caf0f17841d75fef0c"
    "de572cefe4f150846177cd0f19ea4ad"
)

BACKUP_DIR = (
    GOVERNANCE_DIR / "git_recovery"
)

COMMIT_MESSAGE = (
    "Initialize HAT-DSP controlled project scaffold"
)

NOTEBOOK_PATH = (
    "notebooks/"
    "MIVA_MIT_AI_HAT_DSP_MASTER'S_CAPSTONE_PROJECT.ipynb"
)

# ------------------------------------------------------------
# 2. Define verification utilities
# ------------------------------------------------------------

def sha256_bytes(content):
    return hashlib.sha256(content).hexdigest()


def sha256_file(path):
    return sha256_bytes(path.read_bytes())


def read_json(path):
    return json.loads(path.read_text(encoding="utf-8"))


def write_json_exclusive(path, content):

    with path.open("x", encoding="utf-8") as file:
        json.dump(content, file, indent=4)

    assert read_json(path) == content


def git_run(*arguments, check=True, binary=False):

    return subprocess.run(
        ["git", "-C", str(GIT_WORKSPACE), *arguments],
        capture_output=True,
        text=not binary,
        check=check
    )


def git_paths(*arguments):

    output = git_run(
        *arguments,
        binary=True
    ).stdout

    return sorted(
        item.decode("utf-8")
        for item in output.split(b"\x00")
        if item
    )


print("=" * 75)
print("HAT-DSP | D00.3D.2")
print("INITIAL CONTROLLED GIT COMMIT")
print("=" * 75)

# ------------------------------------------------------------
# 3. Verify prerequisite evidence
# ------------------------------------------------------------

for path in [
    CONFIG_PATH,
    MANIFEST_PATH,
    ASSESSMENT_PATH
]:

    assert path.is_file(), f"Missing artifact: {path}"

assert (GIT_WORKSPACE / ".git").is_dir()

assert sha256_file(CONFIG_PATH) == EXPECTED_CONFIG_HASH

PROJECT_CONFIG = read_json(CONFIG_PATH)
SCAFFOLD_MANIFEST = read_json(MANIFEST_PATH)
ASSESSMENT_REPORT = read_json(ASSESSMENT_PATH)

assert ASSESSMENT_REPORT["assessment_status"] == "PASS"
assert ASSESSMENT_REPORT["failed_checks"] == 0
assert ASSESSMENT_REPORT["passed_checks"] == 20
assert ASSESSMENT_REPORT["secret_finding_count"] == 0

assert (
    sha256_file(MANIFEST_PATH)
    == ASSESSMENT_REPORT["scaffold_manifest_sha256"]
)

assert (
    SCAFFOLD_MANIFEST["configuration_sha256"]
    == EXPECTED_CONFIG_HASH
)

assert (
    SCAFFOLD_MANIFEST["scaffold_version"] == "1.0.1"
)

assert (
    PROJECT_CONFIG["clinical_deployment_authorised"]
    is False
)

REGISTER_PATH = (
    GOVERNANCE_DIR
    / ASSESSMENT_REPORT["review_register"]
)

assert REGISTER_PATH.is_file()

assert (
    sha256_file(REGISTER_PATH)
    == ASSESSMENT_REPORT["review_register_sha256"]
)

APPROVED_RECORDS = SCAFFOLD_MANIFEST["scaffold_files"]

APPROVED_FILES = sorted(APPROVED_RECORDS)

assert len(APPROVED_FILES) == 20

assert (
    APPROVED_FILES
    == sorted(ASSESSMENT_REPORT["publication_candidates"])
)

assert (
    git_run("remote", "get-url", "origin").stdout.strip()
    == PROJECT_CONFIG["github_remote_url"]
)

assert (
    git_run("symbolic-ref", "--short", "HEAD").stdout.strip()
    == "main"
)

assert not git_run("ls-files").stdout.strip()
assert not git_run("for-each-ref").stdout.strip()

assert not (GIT_WORKSPACE / NOTEBOOK_PATH).exists()

print("\nPREREQUISITE VERIFICATION")
print("-" * 75)
print("Project configuration integrity : PASS")
print("Publication assessment accepted : PASS")
print("Scaffold manifest integrity     : PASS")
print("Publication register integrity  : PASS")
print("Repository identity             : PASS")
print("Initial repository state        : PASS")

# ------------------------------------------------------------
# 4. Verify publication candidates and exact file contents
# ------------------------------------------------------------

ACTUAL_CANDIDATES = git_paths(
    "ls-files",
    "--others",
    "--exclude-standard",
    "-z"
)

assert ACTUAL_CANDIDATES == APPROVED_FILES

APPROVED_HASHES = {}

for relative_path in APPROVED_FILES:

    path = GIT_WORKSPACE / relative_path

    assert path.is_file()
    assert not path.is_symlink()

    assert path.resolve().is_relative_to(
        GIT_WORKSPACE.resolve()
    )

    record = APPROVED_RECORDS[relative_path]

    assert (
        sha256_file(path) == record["sha256"]
    ), f"Integrity mismatch: {relative_path}"

    assert (
        path.stat().st_size == record["size_bytes"]
    )

    APPROVED_HASHES[relative_path] = record["sha256"]

print("\nAPPROVED COMMIT CANDIDATES")
print("-" * 75)

for relative_path in APPROVED_FILES:
    print(f"{relative_path:<45}: VERIFIED")

# ------------------------------------------------------------
# 5. Obtain explicit local commit authorisation
# ------------------------------------------------------------

print("\nLOCAL COMMIT AUTHORISATION")
print("-" * 75)

print("Approved files       : 20")
print("Automated assessment : PASS")
print("Manual review        : Supplied document changes reviewed")
print("Commit scope         : Local scaffold only")
print("Remote publication   : NOT AUTHORISED")
print("Master notebook      : NOT INCLUDED")

AUTHORISATION = input(
    "\nConfirm your review and type APPROVE LOCAL COMMIT: "
).strip()

if AUTHORISATION != "APPROVE LOCAL COMMIT":

    raise RuntimeError(
        "Local commit not authorised. No files staged."
    )

# ------------------------------------------------------------
# 6. Establish repository-specific Git author identity
# ------------------------------------------------------------

# Use a GitHub-provided no-reply email address to avoid
# exposing your personal email in future public commits.

GIT_AUTHOR_NAME = input(
    "Enter your Git author name: "
).strip()

GIT_AUTHOR_EMAIL = input(
    "Enter your GitHub no-reply email: "
).strip()

assert GIT_AUTHOR_NAME, "Git author name is required."

assert "\n" not in GIT_AUTHOR_NAME
assert "\r" not in GIT_AUTHOR_NAME
assert "<" not in GIT_AUTHOR_NAME
assert ">" not in GIT_AUTHOR_NAME

assert re.fullmatch(
    r"[A-Za-z0-9_.+-]+@users\.noreply\.github\.com",
    GIT_AUTHOR_EMAIL
), "Use your GitHub-provided no-reply email."

git_run("config", "--local", "user.name", GIT_AUTHOR_NAME)

git_run("config", "--local", "user.email", GIT_AUTHOR_EMAIL)

assert (
    git_run(
        "config", "--local", "--get", "user.name"
    ).stdout.strip() == GIT_AUTHOR_NAME
)

assert (
    git_run(
        "config", "--local", "--get", "user.email"
    ).stdout.strip() == GIT_AUTHOR_EMAIL
)

print("\nRepository-specific Git author identity: PASS")

# ------------------------------------------------------------
# 7. Persist authorisation evidence before staging
# ------------------------------------------------------------

RUN_TIMESTAMP = datetime.now(timezone.utc).strftime(
    "%Y%m%dT%H%M%S_%fZ"
)

AUTHORISATION_RECORD = {
    "project": "HAT-DSP",
    "stage": "D00.3D.2",
    "authorised_utc": datetime.now(timezone.utc).isoformat(),
    "configuration_version": "1.0.1",
    "scaffold_version": "1.0.1",
    "scaffold_change_reference": "HAT-DSP-SCR-001",
    "manual_review_confirmed_by_operator": True,
    "local_commit_authorised": True,
    "public_publication_authorised": False,
    "approved_file_count": len(APPROVED_FILES),
    "approved_files": APPROVED_FILES,
    "approved_file_sha256": APPROVED_HASHES,
    "configuration_sha256": EXPECTED_CONFIG_HASH,
    "scaffold_manifest_sha256": sha256_file(MANIFEST_PATH),
    "publication_assessment_sha256": (
        sha256_file(ASSESSMENT_PATH)
    ),
    "commit_message": COMMIT_MESSAGE
}

AUTHORISATION_PATH = (
    GOVERNANCE_DIR
    / f"initial_commit_authorisation_{RUN_TIMESTAMP}.json"
)

write_json_exclusive(
    AUTHORISATION_PATH,
    AUTHORISATION_RECORD
)

# ------------------------------------------------------------
# 8. Stage only approved files
# ------------------------------------------------------------

# Deliberately avoid unrestricted 'git add .' operations.

try:

    assert git_paths(
        "ls-files",
        "--others",
        "--exclude-standard",
        "-z"
    ) == APPROVED_FILES

    for relative_path in APPROVED_FILES:

        assert (
            sha256_file(GIT_WORKSPACE / relative_path)
            == APPROVED_HASHES[relative_path]
        )

    git_run("add", "--", *APPROVED_FILES)

    STAGED_FILES = git_paths(
        "diff",
        "--cached",
        "--name-only",
        "-z"
    )

    assert STAGED_FILES == APPROVED_FILES

    STAGED_HASHES = {}

    for relative_path in STAGED_FILES:

        staged_content = git_run(
            "show",
            f":{relative_path}",
            binary=True
        ).stdout

        staged_hash = sha256_bytes(staged_content)

        assert (
            staged_hash == APPROVED_HASHES[relative_path]
        ), f"Staged content mismatch: {relative_path}"

        STAGED_HASHES[relative_path] = staged_hash

    assert STAGED_HASHES == APPROVED_HASHES

    assert git_run(
        "diff", "--quiet",
        check=False
    ).returncode == 0

except Exception:

    # Restore the initially empty index if staging
    # or staged-content verification fails.

    git_run("read-tree", "--empty")

    raise

print("\nSTAGED CONTENT VERIFICATION")
print("-" * 75)
print("Explicit candidate allowlist : PASS")
print("Staged file count            :", len(STAGED_FILES))
print("Staged SHA-256 integrity     : PASS")
print("Unexpected staged files      : 0")
print("Unstaged modifications       : 0")

# ------------------------------------------------------------
# 9. Establish initial controlled commit
# ------------------------------------------------------------

git_run(
    "commit",
    "-m",
    COMMIT_MESSAGE
)

COMMIT_SHA = git_run(
    "rev-parse", "HEAD"
).stdout.strip()

COMMIT_COUNT = int(
    git_run(
        "rev-list", "--count", "HEAD"
    ).stdout.strip()
)

COMMITTED_FILES = git_paths(
    "ls-tree",
    "-r",
    "--name-only",
    "-z",
    "HEAD"
)

COMMITTED_HASHES = {}

for relative_path in COMMITTED_FILES:

    committed_content = git_run(
        "show",
        f"HEAD:{relative_path}",
        binary=True
    ).stdout

    COMMITTED_HASHES[relative_path] = (
        sha256_bytes(committed_content)
    )

WORKSPACE_CLEAN = (
    not git_run(
        "status",
        "--porcelain",
        "--untracked-files=all"
    ).stdout.strip()
)

GIT_OBJECT_INTEGRITY = (
    git_run(
        "fsck",
        "--full",
        check=False
    ).returncode == 0
)

# ------------------------------------------------------------
# 10. Persist Git recovery bundle
# ------------------------------------------------------------

BACKUP_DIR.mkdir(
    parents=True,
    exist_ok=True
)

BUNDLE_PATH = (
    BACKUP_DIR
    / f"HAT_DSP_initial_commit_{RUN_TIMESTAMP}.bundle"
)

assert not BUNDLE_PATH.exists()

git_run(
    "bundle",
    "create",
    str(BUNDLE_PATH),
    "--all"
)

BUNDLE_VERIFIED = (
    git_run(
        "bundle",
        "verify",
        str(BUNDLE_PATH),
        check=False
    ).returncode == 0
)

BUNDLE_HEADS = git_run(
    "bundle",
    "list-heads",
    str(BUNDLE_PATH)
).stdout.splitlines()

BUNDLE_COMMIT_MATCH = (
    f"{COMMIT_SHA} refs/heads/main" in BUNDLE_HEADS
)

BUNDLE_SHA256 = sha256_file(BUNDLE_PATH)

# ------------------------------------------------------------
# 11. Evaluate final acceptance criteria
# ------------------------------------------------------------

VERIFICATION_RESULTS = {
    "project_configuration_integrity": (
        sha256_file(CONFIG_PATH) == EXPECTED_CONFIG_HASH
    ),

    "approved_manifest_integrity": (
        sha256_file(MANIFEST_PATH)
        == ASSESSMENT_REPORT["scaffold_manifest_sha256"]
    ),

    "publication_assessment_accepted": (
        ASSESSMENT_REPORT["assessment_status"] == "PASS"
        and ASSESSMENT_REPORT["failed_checks"] == 0
    ),

    "manual_commit_authorisation_recorded": (
        read_json(AUTHORISATION_PATH)
        == AUTHORISATION_RECORD
    ),

    "approved_candidate_allowlist": (
        ACTUAL_CANDIDATES == APPROVED_FILES
    ),

    "staged_file_allowlist": (
        STAGED_FILES == APPROVED_FILES
    ),

    "staged_file_hash_integrity": (
        STAGED_HASHES == APPROVED_HASHES
    ),

    "initial_commit_created": (
        bool(re.fullmatch(r"[0-9a-f]{40,64}", COMMIT_SHA))
    ),

    "exactly_one_commit": (
        COMMIT_COUNT == 1
    ),

    "committed_file_allowlist": (
        COMMITTED_FILES == APPROVED_FILES
    ),

    "committed_file_hash_integrity": (
        COMMITTED_HASHES == APPROVED_HASHES
    ),

    "local_branch_main": (
        git_run(
            "branch", "--show-current"
        ).stdout.strip() == "main"
    ),

    "clean_local_workspace": WORKSPACE_CLEAN,

    "git_object_integrity": GIT_OBJECT_INTEGRITY,

    "persistent_recovery_bundle_created": (
        BUNDLE_PATH.is_file()
        and BUNDLE_PATH.stat().st_size > 0
    ),

    "recovery_bundle_verified": BUNDLE_VERIFIED,

    "recovery_bundle_commit_match": BUNDLE_COMMIT_MATCH,

    "remote_configuration_preserved": (
        git_run(
            "remote", "get-url", "origin"
        ).stdout.strip()
        == PROJECT_CONFIG["github_remote_url"]
    ),

    "master_notebook_not_committed": (
        NOTEBOOK_PATH not in COMMITTED_FILES
    ),

    "clinical_deployment_restricted": (
        PROJECT_CONFIG["clinical_deployment_authorised"]
        is False
    )
}

TOTAL_CHECKS = len(VERIFICATION_RESULTS)
PASSED_CHECKS = sum(VERIFICATION_RESULTS.values())
FAILED_CHECKS = TOTAL_CHECKS - PASSED_CHECKS

COMMIT_STATUS = (
    "PASS" if FAILED_CHECKS == 0 else "FAIL"
)

# ------------------------------------------------------------
# 12. Persist commit acceptance evidence
# ------------------------------------------------------------

COMMIT_REPORT = {
    "project": "HAT-DSP",
    "stage": "D00.3D.2",
    "verified_utc": datetime.now(timezone.utc).isoformat(),
    "configuration_version": "1.0.1",
    "scaffold_version": "1.0.1",
    "scaffold_change_reference": "HAT-DSP-SCR-001",
    "configuration_sha256": EXPECTED_CONFIG_HASH,
    "scaffold_manifest_sha256": sha256_file(MANIFEST_PATH),
    "publication_assessment_sha256": (
        sha256_file(ASSESSMENT_PATH)
    ),
    "authorisation_record": AUTHORISATION_PATH.name,
    "authorisation_record_sha256": (
        sha256_file(AUTHORISATION_PATH)
    ),
    "repository_branch": "main",
    "repository_visibility": "public",
    "commit_sha": COMMIT_SHA,
    "commit_message": COMMIT_MESSAGE,
    "commit_count": COMMIT_COUNT,
    "committed_file_count": len(COMMITTED_FILES),
    "committed_files": COMMITTED_FILES,
    "committed_file_sha256": COMMITTED_HASHES,
    "recovery_bundle": BUNDLE_PATH.name,
    "recovery_bundle_sha256": BUNDLE_SHA256,
    "recovery_bundle_verified": BUNDLE_VERIFIED,
    "fresh_runtime_recovery_tested": False,
    "verification_results": VERIFICATION_RESULTS,
    "total_checks": TOTAL_CHECKS,
    "passed_checks": PASSED_CHECKS,
    "failed_checks": FAILED_CHECKS,
    "local_commit_completed": True,
    "master_notebook_committed": False,
    "public_publication_authorised": False,
    "remote_push_performed_by_stage": False,
    "remote_synchronization_verified": False,
    "clinical_deployment_authorised": False,
    "commit_status": COMMIT_STATUS
}

REPORT_PATH = (
    LOG_DIR
    / f"D00_3D_2_initial_commit_{RUN_TIMESTAMP}.json"
)

write_json_exclusive(
    REPORT_PATH,
    COMMIT_REPORT
)

# ------------------------------------------------------------
# 13. Display final acceptance results
# ------------------------------------------------------------

print("\nFINAL COMMIT VERIFICATION")
print("-" * 75)

for check, result in VERIFICATION_RESULTS.items():

    print(
        f"{check.replace('_', ' ').title():<48}: "
        f"{'PASS' if result else 'FAIL'}"
    )

print("\nINITIAL COMMIT SUMMARY")
print("-" * 75)

print("Configuration version        : 1.0.1")
print("Scaffold version             : 1.0.1")
print("Repository branch            : main")
print("Repository visibility        : PUBLIC")
print("Commit count                 :", COMMIT_COUNT)
print("Approved files               :", len(APPROVED_FILES))
print("Committed files              :", len(COMMITTED_FILES))
print("Total verification checks    :", TOTAL_CHECKS)
print("Passed checks                :", PASSED_CHECKS)
print("Failed checks                :", FAILED_CHECKS)
print("Manual commit authorisation  : RECORDED")
print("Recovery bundle              :", (
    "VERIFIED" if BUNDLE_VERIFIED else "FAILED"
))
print("Fresh-runtime recovery test  : PENDING")
print("Master notebook committed    : NO")
print("Remote push                  : NOT PERFORMED")
print("Public publication           : PENDING")
print("Acceptance report integrity  : PASS")

print("\nINITIAL COMMIT SHA")
print(COMMIT_SHA)

print("\nCOMMIT MESSAGE")
print(COMMIT_MESSAGE)

print("\nPERSISTENT GIT RECOVERY BUNDLE")
print(BUNDLE_PATH)

print("\nRECOVERY BUNDLE SHA-256")
print(BUNDLE_SHA256)

print("\nCOMMIT AUTHORISATION RECORD")
print(AUTHORISATION_PATH)

print("\nACCEPTANCE REPORT")
print(REPORT_PATH)

print("=" * 75)
print("D00.3D.2 INITIAL COMMIT STATUS:", COMMIT_STATUS)
print("=" * 75)

assert FAILED_CHECKS == 0, (
    "Initial commit acceptance criteria not satisfied."
)
```

```python

# ============================================================
# HAT-DSP
# D00.4 — REPRODUCIBLE SOURCE-CODE ARCHITECTURE
# ============================================================

from pathlib import Path
from datetime import datetime, timezone

import hashlib
import json
import subprocess

# ------------------------------------------------------------
# 1. Establish authoritative project locations
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    "/content/drive/MyDrive/HAT_DSP_MIVA_MIT_AI_CAPSTONE"
)

GOVERNANCE_DIR = PROJECT_ROOT / "00_project_governance"
SOURCE_ROOT = PROJECT_ROOT / "25_source_code"
LOG_DIR = PROJECT_ROOT / "23_logs"

GIT_WORKSPACE = Path(
    "/content/HAT_DSP_MIVA_MIT_AI_MASTER_CAPSTONE"
)

CONFIG_PATH = GOVERNANCE_DIR / "project_config_v1_0_1.json"

EXPECTED_CONFIG_HASH = (
    "2e9d802885b8759caf0f17841d75fef0c"
    "de572cefe4f150846177cd0f19ea4ad"
)

EXPECTED_COMMIT = (
    "006fdd2783a37a978e8c637109df77c57dcf065f"
)

COMMIT_REPORT_PATH = (
    LOG_DIR
    / "D00_3D_2_initial_commit_20261118T202436_246540Z.json"
)

def sha256_file(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()

def git_run(*args):
    return subprocess.run(
        ["git", "-C", str(GIT_WORKSPACE), *args],
        capture_output=True,
        text=True,
        check=True
    ).stdout.strip()

print("=" * 75)
print("HAT-DSP | D00.4")
print("REPRODUCIBLE SOURCE-CODE ARCHITECTURE")
print("=" * 75)

# ------------------------------------------------------------
# 2. Verify accepted engineering baseline
# ------------------------------------------------------------

assert PROJECT_ROOT.is_dir()
assert SOURCE_ROOT.is_dir()
assert sha256_file(CONFIG_PATH) == EXPECTED_CONFIG_HASH

PROJECT_CONFIG = json.loads(
    CONFIG_PATH.read_text(encoding="utf-8")
)

COMMIT_REPORT = json.loads(
    COMMIT_REPORT_PATH.read_text(encoding="utf-8")
)

assert COMMIT_REPORT["commit_status"] == "PASS"
assert COMMIT_REPORT["commit_sha"] == EXPECTED_COMMIT
assert git_run("rev-parse", "HEAD") == EXPECTED_COMMIT
assert PROJECT_CONFIG["clinical_deployment_authorised"] is False

BUNDLE_PATH = (
    GOVERNANCE_DIR
    / "git_recovery"
    / COMMIT_REPORT["recovery_bundle"]
)

assert (
    sha256_file(BUNDLE_PATH)
    == COMMIT_REPORT["recovery_bundle_sha256"]
)

print("\nENGINEERING BASELINE")
print("-" * 75)
print("Project configuration : VERIFIED")
print("Initial Git commit    : VERIFIED")
print("Git recovery bundle   : VERIFIED")
print("Remote synchronization: DEFERRED")

# ------------------------------------------------------------
# 3. Establish persistent Python package architecture
# ------------------------------------------------------------

PACKAGE_ROOT = SOURCE_ROOT / "hat_dsp"

MODULE_REGISTRY = {
    "ingestion": {
        "purpose": "Source data ingestion and provenance",
        "stages": ["D01"]
    },
    "validation": {
        "purpose": "Data quality and cohort validation",
        "stages": ["D01", "D02", "D03"]
    },
    "preprocessing": {
        "purpose": "Feature governance and preprocessing",
        "stages": ["D04", "D05", "D06", "D07"]
    },
    "modeling": {
        "purpose": "Predictive model development and registry",
        "stages": ["D08", "D13"]
    },
    "evaluation": {
        "purpose": "Performance, clinical utility, fairness and robustness",
        "stages": ["D09", "D10", "D11", "D14"]
    },
    "explainability": {
        "purpose": "Model interpretation and explainability",
        "stages": ["D12"]
    },
    "governance": {
        "purpose": "AI governance, monitoring and risk controls",
        "stages": ["D15", "D17"]
    },
    "assessment": {
        "purpose": "Healthcare technology and investment assessment",
        "stages": ["D16", "D17"]
    },
    "utils": {
        "purpose": "Shared configuration and reproducibility utilities",
        "stages": ["D00-D19"]
    },
    "interface": {
        "purpose": "Integrated Streamlit application services",
        "stages": ["D18"]
    },
    "testing": {
        "purpose": "Functional, integration and acceptance testing",
        "stages": ["D19"]
    }
}

PACKAGE_ROOT.mkdir(exist_ok=True)

EXPECTED_FILES = {
    PACKAGE_ROOT / "__init__.py": (
        '"""HAT-DSP research and prototype package."""\n'
        '__version__ = "0.1.0"\n'
    )
}

for module_name, metadata in MODULE_REGISTRY.items():

    module_path = PACKAGE_ROOT / module_name

    module_path.mkdir(exist_ok=True)

    EXPECTED_FILES[module_path / "__init__.py"] = (
        f'"""{metadata["purpose"]}."""\n'
    )

# Validate all existing files before writing.

for path, content in EXPECTED_FILES.items():

    if path.exists():
        assert path.is_file()
        assert not path.is_symlink()
        assert path.read_text(encoding="utf-8") == content, (
            f"Unexpected existing source file: {path}"
        )

# Create only missing package files.

for path, content in EXPECTED_FILES.items():

    if not path.exists():

        with path.open("x", encoding="utf-8") as file:
            file.write(content)

print("\nPERSISTENT SOURCE-CODE ARCHITECTURE")
print("-" * 75)

for module_name, metadata in MODULE_REGISTRY.items():

    print(f"{module_name:<20}: {metadata['purpose']}")

# ------------------------------------------------------------
# 4. Verify package syntax and imports
# ------------------------------------------------------------

SYNTAX_RESULTS = {}

for path in EXPECTED_FILES:

    relative_path = path.relative_to(SOURCE_ROOT).as_posix()

    try:

        compile(
            path.read_text(encoding="utf-8"),
            str(path),
            "exec"
        )

        SYNTAX_RESULTS[relative_path] = True

    except SyntaxError:

        SYNTAX_RESULTS[relative_path] = False

# Test package imports in a fresh Python process.

import sys
import os

IMPORT_NAMES = [
    "hat_dsp",
    *[
        f"hat_dsp.{module_name}"
        for module_name in MODULE_REGISTRY
    ]
]

IMPORT_SCRIPT = (
    "import importlib\n"
    "import json\n"
    f"modules = {IMPORT_NAMES!r}\n"
    "results = {}\n"
    "for name in modules:\n"
    "    module = importlib.import_module(name)\n"
    "    results[name] = module.__file__\n"
    "print(json.dumps(results))\n"
)

PROCESS_ENV = os.environ.copy()
PROCESS_ENV["PYTHONPATH"] = str(SOURCE_ROOT)
PROCESS_ENV["PYTHONDONTWRITEBYTECODE"] = "1"

IMPORT_PROCESS = subprocess.run(
    [sys.executable, "-c", IMPORT_SCRIPT],
    env=PROCESS_ENV,
    capture_output=True,
    text=True
)

assert IMPORT_PROCESS.returncode == 0, (
    IMPORT_PROCESS.stderr
)

IMPORTED_LOCATIONS = json.loads(
    IMPORT_PROCESS.stdout
)

IMPORT_RESULTS = {
    name: Path(location).resolve().is_relative_to(
        PACKAGE_ROOT.resolve()
    )
    for name, location in IMPORTED_LOCATIONS.items()
}

# ------------------------------------------------------------
# 5. Generate source architecture manifest
# ------------------------------------------------------------

RUN_TIMESTAMP = datetime.now(timezone.utc).strftime(
    "%Y%m%dT%H%M%S_%fZ"
)

SOURCE_MANIFEST = {
    "project": "HAT-DSP",
    "stage": "D00.4",
    "package": "hat_dsp",
    "package_version": "0.1.0",
    "configuration_version": "1.0.1",
    "configuration_sha256": EXPECTED_CONFIG_HASH,
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "modules": MODULE_REGISTRY,
    "source_files": {
        path.relative_to(SOURCE_ROOT).as_posix(): {
            "sha256": sha256_file(path),
            "size_bytes": path.stat().st_size
        }
        for path in EXPECTED_FILES
    },
    "implementation_status": "PACKAGE_SCAFFOLD_ONLY"
}

MANIFEST_PATH = (
    GOVERNANCE_DIR
    / f"source_architecture_manifest_{RUN_TIMESTAMP}.json"
)

with MANIFEST_PATH.open("x", encoding="utf-8") as file:
    json.dump(SOURCE_MANIFEST, file, indent=4)

# ------------------------------------------------------------
# 6. Evaluate acceptance criteria
# ------------------------------------------------------------

VERIFICATION_RESULTS = {

    "project_configuration_integrity": (
        sha256_file(CONFIG_PATH) == EXPECTED_CONFIG_HASH
    ),

    "initial_commit_preserved": (
        git_run("rev-parse", "HEAD") == EXPECTED_COMMIT
    ),

    "persistent_recovery_bundle_integrity": (
        sha256_file(BUNDLE_PATH)
        == COMMIT_REPORT["recovery_bundle_sha256"]
    ),

    "persistent_package_root": PACKAGE_ROOT.is_dir(),

    "module_architecture_complete": all(
        (PACKAGE_ROOT / name).is_dir()
        for name in MODULE_REGISTRY
    ),

    "package_files_complete": all(
        path.is_file()
        for path in EXPECTED_FILES
    ),

    "package_source_integrity": all(
        path.read_text(encoding="utf-8") == content
        for path, content in EXPECTED_FILES.items()
    ),

    "python_syntax_valid": all(
        SYNTAX_RESULTS.values()
    ),

    "fresh_process_imports_valid": (
        set(IMPORT_RESULTS) == set(IMPORT_NAMES)
        and all(IMPORT_RESULTS.values())
    ),

    "source_manifest_integrity": (
        json.loads(MANIFEST_PATH.read_text(encoding="utf-8"))
        == SOURCE_MANIFEST
    ),

    "local_git_workspace_unchanged": (
        not git_run("status", "--porcelain")
    ),

    "clinical_deployment_restricted": (
        PROJECT_CONFIG["clinical_deployment_authorised"]
        is False
    )
}

TOTAL_CHECKS = len(VERIFICATION_RESULTS)
PASSED_CHECKS = sum(VERIFICATION_RESULTS.values())
FAILED_CHECKS = TOTAL_CHECKS - PASSED_CHECKS

STATUS = "PASS" if FAILED_CHECKS == 0 else "FAIL"

# ------------------------------------------------------------
# 7. Persist verification evidence
# ------------------------------------------------------------

REPORT = {
    "project": "HAT-DSP",
    "stage": "D00.4",
    "verified_utc": datetime.now(timezone.utc).isoformat(),
    "configuration_version": "1.0.1",
    "package_version": "0.1.0",
    "module_count": len(MODULE_REGISTRY),
    "source_file_count": len(EXPECTED_FILES),
    "syntax_results": SYNTAX_RESULTS,
    "import_results": IMPORT_RESULTS,
    "source_manifest": MANIFEST_PATH.name,
    "source_manifest_sha256": sha256_file(MANIFEST_PATH),
    "initial_commit_sha": EXPECTED_COMMIT,
    "remote_synchronization_status": "DEFERRED",
    "source_package_committed": False,
    "functional_modules_implemented": False,
    "verification_results": VERIFICATION_RESULTS,
    "total_checks": TOTAL_CHECKS,
    "passed_checks": PASSED_CHECKS,
    "failed_checks": FAILED_CHECKS,
    "architecture_status": STATUS
}

REPORT_PATH = (
    LOG_DIR
    / f"D00_4_source_architecture_{RUN_TIMESTAMP}.json"
)

with REPORT_PATH.open("x", encoding="utf-8") as file:
    json.dump(REPORT, file, indent=4)

assert json.loads(
    REPORT_PATH.read_text(encoding="utf-8")
) == REPORT

# ------------------------------------------------------------
# 8. Display acceptance summary
# ------------------------------------------------------------

print("\nFINAL ARCHITECTURE VERIFICATION")
print("-" * 75)

for check, result in VERIFICATION_RESULTS.items():

    print(
        f"{check.replace('_', ' ').title():<48}: "
        f"{'PASS' if result else 'FAIL'}"
    )

print("\nSOURCE ARCHITECTURE SUMMARY")
print("-" * 75)

print("Python package              : hat_dsp")
print("Package version             : 0.1.0")
print("Functional module directories:", len(MODULE_REGISTRY))
print("Package source files        :", len(EXPECTED_FILES))
print("Syntax checks passed        :", sum(SYNTAX_RESULTS.values()))
print("Import checks passed        :", sum(IMPORT_RESULTS.values()))
print("Total verification checks   :", TOTAL_CHECKS)
print("Passed checks               :", PASSED_CHECKS)
print("Failed checks               :", FAILED_CHECKS)
print("Initial Git commit          : PRESERVED")
print("Remote synchronization      : DEFERRED")
print("Functional implementation   : PENDING")

print("\nPERSISTENT SOURCE PACKAGE")
print(PACKAGE_ROOT)

print("\nSOURCE ARCHITECTURE MANIFEST")
print(MANIFEST_PATH)

print("\nVERIFICATION REPORT")
print(REPORT_PATH)

print("=" * 75)
print("D00.4 SOURCE ARCHITECTURE STATUS:", STATUS)
print("=" * 75)

assert FAILED_CHECKS == 0
```
